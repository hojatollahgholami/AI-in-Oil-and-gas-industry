````python
import pandas as pd
from factor_analyzer import FactorAnalyzer
from sklearn.datasets import load_iris

# بارگیری داده‌های نمونه (Iris)
data = load_iris()
df = pd.DataFrame(data.data, columns=data.feature_names)

# بررسی داده‌های گم‌شده
df.dropna(inplace=True)
````

````python
from factor_analyzer.factor_analyzer import calculate_kmo
kmo_all, kmo_model = calculate_kmo(df)
print(f"KMO Score: {kmo_model:.3f}")  # KMO > 0.6 قابل قبول
````

````python
from factor_analyzer.factor_analyzer import calculate_bartlett_sphericity
chi_sq, p_value = calculate_bartlett_sphericity(df)
print(f"Bartlett's Test p-value: {p_value:.4f}")  # p-value < 0.05
````

````python
import matplotlib.pyplot as plt

fa = FactorAnalyzer(n_factors=df.shape[1], rotation=None)
fa.fit(df)
ev, v = fa.get_eigenvalues()

plt.plot(range(1, df.shape[1]+1), ev, 'o-')
plt.title('Scree Plot')
plt.xlabel('Factors')
plt.ylabel('Eigenvalue')
plt.grid()
plt.show()
````

````python
n_factors = sum(ev > 1)  # تعداد عوامل با مقدار ویژه > 1
print(f"Suggested number of factors: {n_factors}")
````

````python
# تنظیم تعداد عوامل و روش چرخش (Rotation)
fa = FactorAnalyzer(n_factors=n_factors, rotation='varimax')  # چرخش واریماکس
fa.fit(df)

# بارهای عاملی (Factor Loadings)
loadings = pd.DataFrame(fa.loadings_,
                        index=df.columns,
                        columns=[f"Factor_{i}" for i in range(1, n_factors+1)])
print("Factor Loadings:\n", loadings)
````

````python
# نصب کتابخانه‌ها
!pip install factor_analyzer pandas numpy matplotlib seaborn

# ایمپورت کتابخانه‌های ضروری
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from factor_analyzer import FactorAnalyzer
from factor_analyzer.factor_analyzer import calculate_kmo, calculate_bartlett_sphericity

# تولید داده‌های شبیه‌سازی شده (مانند مقاله)
np.random.seed(42)
n_samples = 300
variables = {
    'Leadership': np.random.normal(0, 1, n_samples),
    'Teamwork': np.random.normal(0, 1, n_samples),
    'Creativity': 0.7*np.random.normal(0, 1, n_samples) + 0.3*np.random.normal(0, 1, n_samples),
    'Analytical': 0.6*np.random.normal(0, 1, n_samples) + 0.4*np.random.normal(0, 1, n_samples),
    'Communication': 0.8*np.random.normal(0, 1, n_samples) + 0.2*np.random.normal(0, 1, n_samples),
    'Adaptability': 0.5*np.random.normal(0, 1, n_samples) + 0.5*np.random.normal(0, 1, n_samples)
}

df = pd.DataFrame(variables)
````

````python
plt.figure(figsize=(10, 6))
corr_matrix = df.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('ماتریس همبستگی متغیرها')
plt.show()
````

````python
# آزمون KMO
kmo_all, kmo_model = calculate_kmo(df)
print(f"شاخص KMO: {kmo_model:.3f}")  # > 0.6 قابل قبول

# آزمون بارتلت
chi_sq, p_value = calculate_bartlett_sphericity(df)
print(f"آزمون بارتلت (p-value): {p_value:.5f}")  # < 0.05
````

````python
fa = FactorAnalyzer(n_factors=6, rotation=None)
fa.fit(df)

# محاسبه مقادیر ویژه
ev, v = fa.get_eigenvalues()

# ترسیم نمودار
plt.figure(figsize=(10, 6))
plt.plot(range(1, df.shape[1]+1), ev, 'o--', markersize=8)
plt.axhline(y=1, color='r', linestyle='--')
plt.title('نمودار اسکری (Scree Plot)')
plt.xlabel('شماره عامل')
plt.ylabel('مقدار ویژه (Eigenvalue)')
plt.grid()
plt.show()
````

````python
# پیاده‌سازی آنالیز موازی
n_simulations = 100
ev_simulated = np.zeros((n_simulations, df.shape[1]))

for i in range(n_simulations):
    simulated = np.random.normal(0, 1, size=df.shape)
    fa_sim = FactorAnalyzer(n_factors=df.shape[1], rotation=None)
    fa_sim.fit(simulated)
    ev_sim, _ = fa_sim.get_eigenvalues()
    ev_simulated[i] = ev_sim

# محاسبه صدک 95ام
ev_95 = np.percentile(ev_simulated, 95, axis=0)

# مقایسه با داده‌های واقعی
n_factors = sum(ev > ev_95)
print(f"تعداد عوامل پیشنهادی: {n_factors}")
````

````python
# تنظیم پارامترها
fa = FactorAnalyzer(n_factors=n_factors,
                   rotation='varimax',  # چرخش واریماکس
                   method='minres')     # روش کمترین مربعات

fa.fit(df)

# استخراج بارهای عاملی
loadings = pd.DataFrame(
    fa.loadings_,
    index=df.columns,
    columns=[f"عامل {i+1}" for i in range(n_factors)]
)

# نمایش جدول بارهای عاملی
print("\nبارهای عاملی (Factor Loadings):")
display(loadings.style.background_gradient(cmap='coolwarm', vmin=-1, vmax=1))
````

````python
variance = fa.get_factor_variance()
variance_df = pd.DataFrame({
    'واریانس': variance[0],
    'درصد واریانس': variance[1],
    'درصد تجمعی': variance[2]
}, index=[f"عامل {i+1}" for i in range(n_factors)])

print("\nواریانس تبیین‌شده توسط هر عامل:")
display(variance_df)
````

````python
# محاسبه نمرات عوامل برای هر نمونه
factor_scores = fa.transform(df)

# تبدیل به دیتافریم
scores_df = pd.DataFrame(
    factor_scores,
    columns=[f"Factor_{i+1}_Score" for i in range(n_factors)]
)

# الحاق به داده اصلی
result_df = pd.concat([df, scores_df], axis=1)

# ذخیره‌سازی نتایج
result_df.to_csv('factor_analysis_results.csv', index=False)
````

````python
import pandas as pd
import numpy as np

np.random.seed(170)
n_samples = 200

# تولید داده‌ها با ساختار عاملی پنهان
data = {
    # عامل ۱: استرس مکانیکی (بارگذاری روی لرزش و دما)
    'دمای_کارکرد': 0.8*np.random.normal(70, 5, n_samples) + 0.2*np.random.normal(0,1,n_samples),
    'لرزش_محوری': 0.7*np.random.normal(4.2, 0.5, n_samples) + 0.3*np.random.normal(0,1,n_samples),
    'لرزش_شعاعی': 0.6*np.random.normal(3.8, 0.6, n_samples) + 0.4*np.random.normal(0,1,n_samples),

    # عامل ۲: فشار سیستمی
    'فشار_ورودی': 0.9*np.random.normal(120, 10, n_samples) + 0.1*np.random.normal(0,1,n_samples),
    'فشار_خروجی': 0.8*np.random.normal(85, 7, n_samples) + 0.2*np.random.normal(0,1,n_samples),

    # عامل ۳: خوردگی و فرسودگی
    'غلظت_خوردگی': 0.7*np.random.normal(45, 8, n_samples) + 0.3*np.random.normal(0,1,n_samples),
    'ساعات_کارکرد_پیوسته': np.random.uniform(100, 1000, n_samples),

    # متغیر وابسته (هدف)
    'وضعیت_خرابی': np.random.choice([0,1], size=n_samples, p=[0.7,0.3])
}

df = pd.DataFrame(data)
````

````python
import numpy as np
import matplotlib.pyplot as plt
from pymoo.core.problem import Problem
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize as minimize_moo
from arabic_reshaper import arabic_reshaper
from bidi.algorithm import get_display
from pymoo.config import Config

Config.warnings['not_compiled'] = False

# تنظیمات فارسی‌نویسی
plt.rcParams["font.family"] = "Adobe Arabic"
plt.rcParams["axes.unicode_minus"] = False

def persian_text(text):
    reshaped = arabic_reshaper.reshape(text)
    return get_display(reshaped)

# =============================================================================
# بخش 1: بهینه‌سازی تک‌هدفه با مثال صنعت نفت و گاز
# =============================================================================
def oil_well_cost(x):
    """تابع هزینه حفاری چاه نفت"""
    drilling_cost = 0.5 * np.exp(0.7 * x)
    equipment_cost = 0.3 * x
    penalty = 2.0 * (x - 3.5)**4
    return drilling_cost + equipment_cost + penalty + 2.0

def gradient(x):
    """محاسبه گرادیان به صورت عددی"""
    h = 1e-5
    return (oil_well_cost(x + h) - oil_well_cost(x - h)) / (2 * h)

def gradient_descent(learning_rate=0.01, max_iter=100, x0=1.0):
    """الگوریتم گرادیان کاهشی"""
    x = x0
    history = [x]
    costs = [oil_well_cost(x)]

    for _ in range(max_iter):
        grad = gradient(x)
        x = x - learning_rate * grad
        history.append(x)
        costs.append(oil_well_cost(x))

    return np.array(history), np.array(costs)

# اجرای الگوریتم
gd_path, gd_costs = gradient_descent(learning_rate=0.05, max_iter=50, x0=1.0)
optimal_depth = gd_path[-1]

# =============================================================================
# بخش 2: بهینه‌سازی چندهدفه در اکتشاف نفت
# =============================================================================
class OilExplorationProblem(Problem):
    def __init__(self):
        """مسئله بهینه‌سازی چندهدفه برای اکتشاف نفت"""
        super().__init__(n_var=2, n_obj=3, n_constr=0, xl=0.0, xu=10.0)

    def _evaluate(self, X, out, *args, **kwargs):
        x1, x2 = X[:, 0], X[:, 1]

        # تابع هدف 1: هزینه حفاری
        f1 = 0.4 * x2 + 0.1 * (x1 - 5)**2 + 0.5 * np.exp(0.3 * x2)

        # تابع هدف 2: خطر محیط زیستی
        sensitive_area1 = np.sqrt((x1 - 2)**2 + (x2 - 8)**2)
        sensitive_area2 = np.sqrt((x1 - 8)**2 + (x2 - 2)**2)
        f2 = 0.5 * np.exp(-0.5 * sensitive_area1) + 0.5 * np.exp(-0.5 * sensitive_area2)

        # تابع هدف 3: فاصله از مرکز مخزن نفتی
        reservoir_center = (5, 5)
        f3 = np.sqrt((x1 - reservoir_center[0])**2 + (x2 - reservoir_center[1])**2)

        out["F"] = np.column_stack([f1, f2, f3])

# ایجاد و حل مسئله
problem = OilExplorationProblem()
algorithm = NSGA2(pop_size=50)
res = minimize_moo(problem, algorithm, ('n_gen', 30), seed=1, verbose=False)

# استخراج نتایج
pareto_points = res.X
pareto_front = res.F

# =============================================================================
# ترسیم نمودارهای دو بعدی
# =============================================================================
fig, axs = plt.subplots(2, 2, figsize=(16, 12))

# نمودار 1: بهینه‌سازی هزینه حفاری چاه
axs[0, 0].plot(np.linspace(1, 6, 500),
             [oil_well_cost(x) for x in np.linspace(1, 6, 500)],
             'b-', linewidth=2, label=persian_text('تابع هزینه حفاری'))
axs[0, 0].plot(gd_path, gd_costs, 'ro--', alpha=0.7,
              label=persian_text('مسیر بهینه‌سازی (گرادیان کاهشی)'))
axs[0, 0].axvline(optimal_depth, color='g', linestyle='--', alpha=0.7,
                 label=persian_text(f'عمق بهینه: {optimal_depth:.2f} کیلومتر'))
axs[0, 0].set_xlabel(persian_text('عمق چاه (کیلومتر)'), fontsize=12)
axs[0, 0].set_ylabel(persian_text('هزینه (میلیون دلار)'), fontsize=12)
axs[0, 0].set_title(persian_text('بهینه‌سازی عمق حفاری چاه نفت'), fontsize=14)
axs[0, 0].legend()
axs[0, 0].grid(alpha=0.2)
axs[0, 0].annotate(persian_text('هزینه تجهیزات\nافزایش خطی'),
                  xy=(4.5, oil_well_cost(4.5)), xytext=(3.5, 12),
                  arrowprops=dict(arrowstyle='->'))
axs[0, 0].annotate(persian_text('هزینه حفاری\nافزایش نمایی'),
                  xy=(5.5, oil_well_cost(5.5)), xytext=(4.5, 18),
                  arrowprops=dict(arrowstyle='->'))

# نمودار 2: جبهه پارتو دو بعدی (هزینه حفاری در مقابل خطر محیط زیستی)
sc1 = axs[0, 1].scatter(pareto_front[:, 0], pareto_front[:, 1],
                       c=pareto_front[:, 2], cmap='viridis', s=50)
axs[0, 1].set_xlabel(persian_text('هزینه حفاری'), fontsize=12)
axs[0, 1].set_ylabel(persian_text('خطر محیط زیستی'), fontsize=12)
axs[0, 1].set_title(persian_text('جبهه پارتو: هزینه حفاری در مقابل خطر محیط زیستی'), fontsize=14)
axs[0, 1].grid(alpha=0.2)
cbar1 = plt.colorbar(sc1, ax=axs[0, 1])
cbar1.set_label(persian_text('فاصله از مخزن'), fontsize=10)

# نمودار 3: موقعیت‌های بهینه چاه‌ها
scatter = axs[1, 0].scatter(pareto_points[:, 0], pareto_points[:, 1],
                          c=pareto_front[:, 0], cmap='plasma', s=60,
                          label=persian_text('موقعیت‌های بهینه'))

# مناطق حساس
sensitive1 = plt.Circle((2, 8), 1.0, color='red', alpha=0.2, label=persian_text('مناطق حساس محیط زیستی'))
sensitive2 = plt.Circle((8, 2), 1.0, color='red', alpha=0.2)
axs[1, 0].add_patch(sensitive1)
axs[1, 0].add_patch(sensitive2)

# مخزن نفتی
reservoir = plt.Circle((5, 5), 1.5, color='blue', alpha=0.1, label=persian_text('مخزن نفتی'))
axs[1, 0].add_patch(reservoir)

axs[1, 0].set_xlim(0, 10)
axs[1, 0].set_ylim(0, 10)
axs[1, 0].set_xlabel(persian_text('موقعیت افقی (کیلومتر)'), fontsize=12)
axs[1, 0].set_ylabel(persian_text('موقعیت عمودی (کیلومتر)'), fontsize=12)
axs[1, 0].set_title(persian_text('موقعیت‌های بهینه چاه‌های نفت'), fontsize=14)
axs[1, 0].grid(alpha=0.2)
axs[1, 0].legend(loc='upper right')

# نمودار 4: تجزیه و تحلیل مبادله بین اهداف
sc2 = axs[1, 1].scatter(pareto_front[:, 0], pareto_front[:, 2],
                       c=pareto_front[:, 1], cmap='coolwarm', s=50)
axs[1, 1].set_xlabel(persian_text('هزینه حفاری'), fontsize=12)
axs[1, 1].set_ylabel(persian_text('فاصله از مخزن'), fontsize=12)
axs[1, 1].set_title(persian_text('تجزیه و تحلیل مبادله بین اهداف'), fontsize=14)
axs[1, 1].grid(alpha=0.2)
cbar2 = plt.colorbar(sc2, ax=axs[1, 1])
cbar2.set_label(persian_text('خطر محیط زیستی'), fontsize=10)

# توضیحات روی نمودار
axs[1, 1].annotate(persian_text('راه‌حل‌های کم‌هزینه\nو نزدیک به مخزن'),
                  xy=(np.min(pareto_front[:, 0]), np.min(pareto_front[:, 2])),
                  xytext=(1.5, 0.5), arrowprops=dict(arrowstyle='->'))
axs[1, 1].annotate(persian_text('راه‌حل‌های پر هزینه\nاما دور از مناطق حساس'),
                  xy=(np.max(pareto_front[:, 0]), np.max(pareto_front[:, 2])),
                  xytext=(8, 7), arrowprops=dict(arrowstyle='->'))

# تنظیم نهایی
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.suptitle(persian_text('بهینه‌سازی عملیات اکتشاف و تولید نفت و گاز'), fontsize=20)
plt.savefig('oil_gas_optimization_2d.png', dpi=300, bbox_inches='tight')
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from pymoo.core.problem import Problem
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize
from pymoo.config import Config

Config.warnings['not_compiled'] = False

# تابع هزینه حفاری چاه نفت
def oil_well_cost(x):
    drilling_cost = 0.5 * np.exp(0.7 * x)
    equipment_cost = 0.3 * x
    penalty = 2.0 * (x - 3.5)**4
    return drilling_cost + equipment_cost + penalty + 2.0

# مشتقات عددی
def gradient(x):
    h = 1e-5
    return (oil_well_cost(x + h) - oil_well_cost(x - h)) / (2 * h)

def hessian(x):
    h = 1e-5
    return (oil_well_cost(x + h) - 2 * oil_well_cost(x) + oil_well_cost(x - h)) / (h ** 2)

# گرادیان کاهشی
def gradient_descent(x0=1.0, lr=0.05, max_iter=50):
    x = x0
    path = [x]
    for _ in range(max_iter):
        x -= lr * gradient(x)
        path.append(x)
    return np.array(path)

# روش نیوتن
def newton_method(x0=1.0, max_iter=10):
    x = x0
    path = [x]
    for _ in range(max_iter):
        g = gradient(x)
        h = hessian(x)
        if h == 0:
            break
        x -= g / h
        path.append(x)
    return np.array(path)

# مسئله برای الگوریتم ژنتیک
class OilDrillingProblem(Problem):
    def __init__(self):
        super().__init__(n_var=1, n_obj=1, n_constr=0, xl=1.0, xu=6.0)

    def _evaluate(self, X, out, *args, **kwargs):
        out["F"] = oil_well_cost(X[:, 0]).reshape(-1, 1)

problem = OilDrillingProblem()
algorithm = NSGA2(pop_size=40)
result = minimize(problem, algorithm, ('n_gen', 30), verbose=False, seed=1)

# ترسیم
x_vals = np.linspace(1, 6, 400)
y_vals = oil_well_cost(x_vals)

gd_path = gradient_descent()
newton_path = newton_method()
ga_points = result.X.flatten()
ga_scores = result.F.flatten()

plt.figure(figsize=(12, 7))
plt.plot(x_vals, y_vals, label="تابع هزینه حفاری", linewidth=2)
plt.plot(gd_path, oil_well_cost(gd_path), 'ro--', label="گرادیان کاهشی")
plt.plot(newton_path, oil_well_cost(newton_path), 'gs--', label="روش نیوتن")
plt.scatter(ga_points, ga_scores, c='purple', label="الگوریتم ژنتیک", alpha=0.6)

plt.xlabel("عمق چاه (کیلومتر)")
plt.ylabel("هزینه کل (میلیون دلار)")
plt.title("مقایسه الگوریتم‌های بهینه‌سازی برای هزینه حفاری چاه نفت")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm
from pymoo.core.problem import Problem
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize as minimize_moo
from arabic_reshaper import arabic_reshaper
from bidi.algorithm import get_display
from pymoo.config import Config
import matplotlib.gridspec as gridspec

Config.warnings['not_compiled'] = False

# تنظیمات فارسی‌نویسی
plt.rcParams["font.family"] = "Adobe Arabic"
plt.rcParams["axes.unicode_minus"] = False

def persian_text(text):
    reshaped = arabic_reshaper.reshape(text)
    return get_display(reshaped)

# =============================================================================
# مسئله بهینه‌سازی سه‌بعدی برای تاسیسات ذخیره‌سازی گاز
# =============================================================================
class GasStorageOptimization(Problem):
    def __init__(self):
        """
        بهینه‌سازی محل احداث تاسیسات ذخیره‌سازی گاز طبیعی
        متغیرهای تصمیم:
          x: موقعیت طول جغرافیایی (کیلومتر)
          y: موقعیت عرض جغرافیایی (کیلومتر)
          z: عمق مخزن (کیلومتر)

        اهداف:
          1. حداقل‌سازی هزینه احداث
          2. حداقل‌سازی فاصله از خط لوله اصلی
          3. حداکثرسازی ظرفیت ذخیره‌سازی
        """
        super().__init__(n_var=3, n_obj=3, n_constr=0,
                         xl=np.array([0, 0, 0.5]),
                         xu=np.array([10, 10, 3.0]))

    def _evaluate(self, X, out, *args, **kwargs):
        x, y, z = X[:, 0], X[:, 1], X[:, 2]

        # تابع هدف 1: هزینه احداث (ترکیب عمق و موقعیت)
        # هزینه با عمق رابطه نمایی دارد و با فاصله از شهر رابطه خطی
        urban_center = (5, 5)
        dist_to_city = np.sqrt((x - urban_center[0])**2 + (y - urban_center[1])**2)
        f1 = 500 * np.exp(0.8 * z) + 50 * dist_to_city

        # تابع هدف 2: فاصله از خط لوله اصلی (در امتداد y=2)
        pipeline_distance = np.abs(y - 2)
        f2 = pipeline_distance

        # تابع هدف 3: ظرفیت ذخیره‌سازی (منفی می‌کنیم چون می‌خواهیم ماکزیمم شود)
        # ظرفیت با کیفیت زمین و عمق رابطه دارد
        soil_quality = np.exp(-0.2*((x-3)**2 + (y-7)**2)) + 0.7*np.exp(-0.1*((x-8)**2 + (y-3)**2))
        f3 = - (1000 * soil_quality * (3.0 - z))  # منفی برای ماکزیمم سازی

        out["F"] = np.column_stack([f1, f2, f3])

# ایجاد و حل مسئله
problem = GasStorageOptimization()
algorithm = NSGA2(pop_size=100)
res = minimize_moo(problem, algorithm, ('n_gen', 50), seed=1, verbose=False)

# استخراج نتایج
solutions = res.X
objectives = res.F

# جداسازی مقادیر
x_pos = solutions[:, 0]
y_pos = solutions[:, 1]
z_depth = solutions[:, 2]
costs = objectives[:, 0]
pipeline_dist = objectives[:, 1]
storage_cap = -objectives[:, 2]  # تبدیل به مثبت برای ظرفیت واقعی

# =============================================================================
# ترسیم نمودارهای سه‌بعدی
# =============================================================================
fig = plt.figure(figsize=(18, 12))
fig.suptitle(persian_text('بهینه‌سازی سه‌بعدی محل احداث تاسیسات ذخیره‌سازی گاز'), fontsize=20, y=0.98)

# ساختار GridSpec برای چیدمان نمودارها
gs = gridspec.GridSpec(2, 2, width_ratios=[1.2, 1], height_ratios=[1, 1])

# نمودار 1: موقعیت جغرافیایی و عمق مخازن
ax1 = fig.add_subplot(gs[0], projection='3d')
sc1 = ax1.scatter(x_pos, y_pos, z_depth, c=storage_cap, cmap='viridis', s=50, alpha=0.8)

# خط لوله اصلی
y_pipeline = np.full(100, 2)
x_pipeline = np.linspace(0, 10, 100)
z_pipeline = np.linspace(0.5, 3, 100)
ax1.plot(x_pipeline, y_pipeline, z_pipeline, 'r-', linewidth=3, label=persian_text('خط لوله اصلی'))

# مرکز شهر
ax1.scatter([5], [5], [0.5], c='r', s=200, marker='*', label=persian_text('مرکز شهر'))

ax1.set_xlabel(persian_text('موقعیت طول جغرافیایی (کیلومتر)'), fontsize=10, labelpad=15)
ax1.set_ylabel(persian_text('موقعیت عرض جغرافیایی (کیلومتر)'), fontsize=10, labelpad=15)
ax1.set_zlabel(persian_text('عمق مخزن (کیلومتر)'), fontsize=10, labelpad=15)
ax1.set_title(persian_text('موقعیت و عمق بهینه تاسیسات'), fontsize=14)
ax1.legend()

# نمودار 2: هزینه احداث در مقابل فاصله از خط لوله
ax2 = fig.add_subplot(gs[1])
sc2 = ax2.scatter(costs, pipeline_dist, c=storage_cap, cmap='plasma', s=60)

# خطوط راهنما
ax2.axvline(np.median(costs), color='r', linestyle='--', alpha=0.5)
ax2.axhline(np.median(pipeline_dist), color='g', linestyle='--', alpha=0.5)

ax2.set_xlabel(persian_text('هزینه احداث (میلیون دلار)'), fontsize=12)
ax2.set_ylabel(persian_text('فاصله از خط لوله (کیلومتر)'), fontsize=12)
ax2.set_title(persian_text('هزینه احداث در مقابل فاصله از خط لوله'), fontsize=14)
ax2.grid(alpha=0.3)

cbar2 = plt.colorbar(sc2, ax=ax2)
cbar2.set_label(persian_text('ظرفیت ذخیره‌سازی (MCM)'), fontsize=10)

# نمودار 3: ظرفیت ذخیره‌سازی بر حسب عمق و موقعیت
ax3 = fig.add_subplot(gs[2], projection='3d')
sc3 = ax3.scatter(x_pos, y_pos, storage_cap, c=z_depth, cmap='coolwarm', s=50)

# سطح ظرفیت
x_surf, y_surf = np.meshgrid(np.linspace(0, 10, 20), np.linspace(0, 10, 20))
soil_quality = np.exp(-0.2*((x_surf-3)**2 + (y_surf-7)**2)) + 0.7*np.exp(-0.1*((x_surf-8)**2 + (y_surf-3)**2))
storage_surf = 1000 * soil_quality * 2.5  # عمق متوسط 2.5 کیلومتر
ax3.plot_surface(x_surf, y_surf, storage_surf, cmap=cm.coolwarm, alpha=0.3)

ax3.set_xlabel(persian_text('طول جغرافیایی (کیلومتر)'), fontsize=10, labelpad=15)
ax3.set_ylabel(persian_text('عرض جغرافیایی (کیلومتر)'), fontsize=10, labelpad=15)
ax3.set_zlabel(persian_text('ظرفیت ذخیره‌سازی (MCM)'), fontsize=10, labelpad=15)
ax3.set_title(persian_text('ظرفیت ذخیره‌سازی بر حسب موقعیت'), fontsize=14)
ax3.view_init(elev=30, azim=45)

# نمودار 4: تحلیل مبادلات بین اهداف
ax4 = fig.add_subplot(gs[3])

# سه معیار اصلی
metrics = [costs, pipeline_dist, storage_cap]
metric_names = [persian_text('هزینه احداث'),
               persian_text('فاصله از خط لوله'),
               persian_text('ظرفیت ذخیره‌سازی')]

# انتخاب 5 راه‌حل نمونه
sample_indices = np.linspace(0, len(solutions)-1, 5, dtype=int)
colors = plt.cm.viridis(np.linspace(0, 1, 5))

for i, idx in enumerate(sample_indices):
    normalized_metrics = [metrics[0][idx]/np.max(metrics[0]),
                         metrics[1][idx]/np.max(metrics[1]),
                         metrics[2][idx]/np.max(metrics[2])]

    ax4.plot(metric_names, normalized_metrics, 'o-',
            color=colors[i],
            label=persian_text(f'راه‌حل {i+1}'))

ax4.set_title(persian_text('مقایسه راه‌حل‌های بهینه (نرمال‌شده)'), fontsize=14)
ax4.set_ylabel(persian_text('مقدار نرمال‌شده'), fontsize=12)
ax4.legend(loc='best')
ax4.grid(alpha=0.3)
plt.xticks(rotation=15)

# تنظیم فاصله‌ها
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.subplots_adjust(wspace=0.25, hspace=0.3)

plt.savefig('gas_storage_optimization_3d.png', dpi=300, bbox_inches='tight')
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from pymoo.core.problem import Problem
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize
from matplotlib.patches import Circle

# تعریف مسئله بهینه‌سازی چندهدفه
class OilFieldProblem(Problem):
    def __init__(self):
        super().__init__(n_var=2, n_obj=3, n_constr=0, xl=0.0, xu=10.0)

    def _evaluate(self, X, out, *args, **kwargs):
        x1, x2 = X[:, 0], X[:, 1]

        # تابع هدف 1: هزینه حفاری
        f1 = 0.3 * x2 + 0.1 * (x1 - 5)**2 + 0.5 * np.exp(0.3 * x2)

        # تابع هدف 2: خطر محیط زیستی
        sensitive1 = np.sqrt((x1 - 2)**2 + (x2 - 8)**2)
        sensitive2 = np.sqrt((x1 - 8)**2 + (x2 - 2)**2)
        f2 = np.exp(-0.6 * sensitive1) + np.exp(-0.6 * sensitive2)

        # تابع هدف 3: فاصله از مخزن
        reservoir = (5, 5)
        f3 = np.sqrt((x1 - reservoir[0])**2 + (x2 - reservoir[1])**2)

        out["F"] = np.column_stack([f1, f2, f3])

# اجرای NSGA-II
problem = OilFieldProblem()
algorithm = NSGA2(pop_size=60)
result = minimize(problem, algorithm, ('n_gen', 40), seed=1, verbose=False)

# داده‌ها
X = result.X
F = result.F

# ترسیم موقعیت‌های بهینه روی نقشه
fig, ax = plt.subplots(figsize=(8, 8))

sc = ax.scatter(X[:, 0], X[:, 1], c=F[:, 0], cmap='viridis', s=60)
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_xlabel("مختصات افقی (km)")
ax.set_ylabel("مختصات عمودی (km)")
ax.set_title("موقعیت‌های بهینه چاه‌ها بر اساس هزینه حفاری")

# اضافه‌کردن مناطق حساس
ax.add_patch(Circle((2, 8), 1.0, color='red', alpha=0.2, label='منطقه حساس 1'))
ax.add_patch(Circle((8, 2), 1.0, color='red', alpha=0.2, label='منطقه حساس 2'))

# اضافه کردن مخزن نفتی
ax.add_patch(Circle((5, 5), 1.5, color='blue', alpha=0.1, label='مخزن نفت'))

# نمایش رنگ هزینه‌ها
cbar = plt.colorbar(sc)
cbar.set_label("هزینه حفاری")

ax.legend(loc='upper right')
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from pymoo.core.problem import Problem
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize as minimize_moo
from arabic_reshaper import arabic_reshaper
from bidi.algorithm import get_display

# تنظیمات فارسی‌نویسی
plt.rcParams["font.family"] = "Adobe Arabic"
plt.rcParams["axes.unicode_minus"] = False

def persian_text(text):
    reshaped = arabic_reshaper.reshape(text)
    return get_display(reshaped)

# =============================================================================
# مسئله بهینه‌سازی انتخاب تجهیزات خط لوله گاز
# =============================================================================
class GasPipelineOptimization(Problem):
    def __init__(self):
        """
        بهینه‌سازی انتخاب تجهیزات برای خط لوله گاز
        متغیرهای تصمیم:
          x: قطر لوله (اینچ)
          y: فشار کاری (بار)
          z: ضخامت دیواره لوله (میلی‌متر)

        اهداف:
          1. حداقل‌سازی هزینه ساخت
          2. حداقل‌سازی تلفات انرژی
          3. حداکثرسازی ایمنی
        """
        super().__init__(n_var=3, n_obj=3, n_constr=0,
                         xl=np.array([10, 20, 5]),
                         xu=np.array([50, 100, 30]))

    def _evaluate(self, X, out, *args, **kwargs):
        diameter, pressure, thickness = X[:, 0], X[:, 1], X[:, 2]

        # تابع هدف 1: هزینه ساخت
        material_cost = 0.05 * diameter * thickness
        installation_cost = 0.02 * diameter**2
        f1 = material_cost + installation_cost

        # تابع هدف 2: تلفات انرژی
        # تلفات انرژی با فشار رابطه معکوس و با قطر رابطه مستقیم دارد
        energy_loss = 1000 / (pressure * np.sqrt(diameter))
        f2 = energy_loss

        # تابع هدف 3: ایمنی (منفی می‌کنیم چون می‌خواهیم ماکزیمم شود)
        # ایمنی با ضخامت رابطه مستقیم و با فشار رابطه معکوس دارد
        safety = (thickness / 5) * (100 / pressure)
        f3 = -safety  # منفی برای ماکزیمم سازی

        out["F"] = np.column_stack([f1, f2, f3])

# ایجاد و حل مسئله
problem = GasPipelineOptimization()
algorithm = NSGA2(pop_size=100)
res = minimize_moo(problem, algorithm, ('n_gen', 50), seed=1, verbose=False)

# استخراج نتایج
solutions = res.X
objectives = res.F

# جداسازی مقادیر
diameters = solutions[:, 0]
pressures = solutions[:, 1]
thicknesses = solutions[:, 2]
costs = objectives[:, 0]
energy_losses = objectives[:, 1]
safety = -objectives[:, 2]  # تبدیل به مثبت برای ایمنی واقعی

# =============================================================================
# ترسیم نمودارهای شفاف
# =============================================================================
fig = plt.figure(figsize=(18, 12))
fig.suptitle(persian_text('بهینه‌سازی تجهیزات خط لوله گاز'), fontsize=22, y=0.98)

# ساختار GridSpec برای چیدمان نمودارها
gs = gridspec.GridSpec(2, 2, width_ratios=[1, 1.2], height_ratios=[1, 1])

# نمودار 1: رابطه بین قطر لوله و فشار کاری
ax1 = fig.add_subplot(gs[0])
sc1 = ax1.scatter(diameters, pressures, c=costs, cmap='viridis', s=80, alpha=0.8)

# خطوط راهنما برای محدوده‌های ایمنی
ax1.axvline(20, color='r', linestyle='--', alpha=0.4)
ax1.axvline(40, color='r', linestyle='--', alpha=0.4)
ax1.axhline(40, color='g', linestyle='--', alpha=0.4)
ax1.axhline(80, color='g', linestyle='--', alpha=0.4)

ax1.set_xlabel(persian_text('قطر لوله (اینچ)'), fontsize=12)
ax1.set_ylabel(persian_text('فشار کاری (بار)'), fontsize=12)
ax1.set_title(persian_text('رابطه قطر لوله و فشار کاری با رنگ نشان‌دهنده هزینه'), fontsize=14)
ax1.grid(alpha=0.3)

cbar1 = plt.colorbar(sc1, ax=ax1)
cbar1.set_label(persian_text('هزینه ساخت (هزار دلار)'), fontsize=10)

# توضیحات مناطق
ax1.annotate(persian_text('منطقه ایمن'),
            xy=(30, 60), xytext=(15, 90),
            arrowprops=dict(arrowstyle='->'))
ax1.annotate(persian_text('هزینه بالا'),
            xy=(45, 90), xytext=(35, 95),
            arrowprops=dict(arrowstyle='->'))

# نمودار 2: رابطه بین ایمنی و تلفات انرژی
ax2 = fig.add_subplot(gs[1])
sc2 = ax2.scatter(safety, energy_losses, c=thicknesses, cmap='plasma', s=80)

# خط روند
z = np.polyfit(safety, energy_losses, 1)
p = np.poly1d(z)
ax2.plot(safety, p(safety), 'r--', alpha=0.7)

ax2.set_xlabel(persian_text('سطح ایمنی'), fontsize=12)
ax2.set_ylabel(persian_text('تلفات انرژی (ک.و.س)'), fontsize=12)
ax2.set_title(persian_text('مبادله ایمنی و تلفات انرژی'), fontsize=14)
ax2.grid(alpha=0.3)

cbar2 = plt.colorbar(sc2, ax=ax2)
cbar2.set_label(persian_text('ضخامت دیواره (میلی‌متر)'), fontsize=10)

# نمودار 3: تاثیر ضخامت دیواره بر هزینه و ایمنی
ax3 = fig.add_subplot(gs[2])
sc3 = ax3.scatter(thicknesses, costs, c=safety, cmap='coolwarm', s=80)

# تقسیم بندی مناطق
ax3.fill_between([5, 15], 0, 15, color='green', alpha=0.1)
ax3.fill_between([15, 30], 0, 15, color='yellow', alpha=0.1)
ax3.fill_between([25, 30], 0, 15, color='red', alpha=0.1)

ax3.set_xlabel(persian_text('ضخامت دیواره (میلی‌متر)'), fontsize=12)
ax3.set_ylabel(persian_text('هزینه ساخت (هزار دلار)'), fontsize=12)
ax3.set_title(persian_text('تاثیر ضخامت دیواره بر هزینه و ایمنی'), fontsize=14)
ax3.grid(alpha=0.3)

cbar3 = plt.colorbar(sc3, ax=ax3)
cbar3.set_label(persian_text('سطح ایمنی'), fontsize=10)

# توضیحات مناطق
ax3.text(8, 3, persian_text('منطقه بهینه'), fontsize=12, color='green')
ax3.text(20, 3, persian_text('منطقه قابل قبول'), fontsize=12, color='orange')
ax3.text(27, 3, persian_text('هزینه بالا'), fontsize=12, color='red')

# نمودار 4: جبهه پارتو سه‌بعدی
ax4 = fig.add_subplot(gs[3], projection='3d')

# نقاط جبهه پارتو
sc4 = ax4.scatter(costs, energy_losses, safety,
                 c=thicknesses, cmap='viridis', s=50, alpha=0.8)

# برچسب‌های محورها
ax4.set_xlabel(persian_text('هزینه ساخت'), fontsize=10, labelpad=15)
ax4.set_ylabel(persian_text('تلفات انرژی'), fontsize=10, labelpad=15)
ax4.set_zlabel(persian_text('سطح ایمنی'), fontsize=10, labelpad=15)
ax4.set_title(persian_text('جبهه پارتو سه‌بعدی اهداف بهینه‌سازی'), fontsize=14)

# جهت‌های بهینه‌سازی
ax4.quiver(0, 0, 0, 15, 0, 0, color='r', arrow_length_ratio=0.1, label=persian_text('کاهش هزینه'))
ax4.quiver(0, 0, 0, 0, 100, 0, color='g', arrow_length_ratio=0.1, label=persian_text('کاهش تلفات'))
ax4.quiver(0, 0, 0, 0, 0, 100, color='b', arrow_length_ratio=0.1, label=persian_text('افزایش ایمنی'))

ax4.legend()
ax4.view_init(elev=25, azim=45)

# تنظیم فاصله‌ها
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.subplots_adjust(wspace=0.25, hspace=0.3)

plt.savefig('gas_pipeline_optimization.png', dpi=300, bbox_inches='tight')
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from scipy.optimize import minimize
from pymoo.core.problem import Problem
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize as minimize_moo
from arabic_reshaper import arabic_reshaper
from bidi.algorithm import get_display

# تنظیمات فارسی‌نویسی
plt.rcParams["font.family"] = "Adobe Arabic"
plt.rcParams["axes.unicode_minus"] = False

def persian_text(text):
    reshaped = arabic_reshaper.reshape(text)
    return get_display(reshaped)

# =============================================================================
# تابع هزینه خط لوله گاز (بهینه‌سازی تک‌هدفه)
# =============================================================================
def gas_pipeline_cost(D, L=1000):
    """
    تابع هزینه خط لوله گاز بر حسب قطر لوله (D)
    - هزینه مواد: متناسب با سطح مقطع لوله (D^2)
    - هزینه نصب: متناسب با طول لوله (L)
    - هزینه انرژی: متناسب با معکوس توان چهارم قطر (1/D^4)
    """
    material_cost = 0.02 * D**2 * L
    installation_cost = 50 * L
    energy_cost = 10000 / D**4
    return material_cost + installation_cost + energy_cost

def gradient(D, L=1000):
    """محاسبه گرادیان تابع هزینه به صورت تحلیلی"""
    h = 1e-5
    return (gas_pipeline_cost(D + h, L) - gas_pipeline_cost(D - h, L)) / (2 * h)

def hessian(D, L=1000):
    """محاسبه هسیان تابع هزینه به صورت تحلیلی"""
    h = 1e-5
    return (gas_pipeline_cost(D + h, L) - 2*gas_pipeline_cost(D, L) + gas_pipeline_cost(D - h, L)) / h**2

# گرادیان کاهشی
def gradient_descent(learning_rate=0.1, max_iter=100, D0=10.0, L=1000):
    D = D0
    history = [D]
    costs = [gas_pipeline_cost(D, L)]

    for _ in range(max_iter):
        grad = gradient(D, L)
        D = D - learning_rate * grad
        # محدود کردن به بازه معقول
        D = max(5, min(50, D))
        history.append(D)
        costs.append(gas_pipeline_cost(D, L))

    return np.array(history), np.array(costs)

# روش نیوتن
def newton_method(max_iter=10, D0=10.0, L=1000):
    D = D0
    history = [D]
    costs = [gas_pipeline_cost(D, L)]

    for _ in range(max_iter):
        grad = gradient(D, L)
        hess = hessian(D, L)
        if abs(hess) > 1e-5:
            D = D - grad/hess
        # محدود کردن به بازه معقول
        D = max(5, min(50, D))
        history.append(D)
        costs.append(gas_pipeline_cost(D, L))

    return np.array(history), np.array(costs)

# اجرای الگوریتم‌های تک‌هدفه
D0 = 10.0  # نقطه شروع (قطر کوچک)
L_fixed = 1000  # طول ثابت خط لوله

gd_path, gd_costs = gradient_descent(learning_rate=0.5, max_iter=50, D0=D0, L=L_fixed)
newton_path, newton_costs = newton_method(max_iter=10, D0=D0, L=L_fixed)

# نقطه بهینه
optimal_D = gd_path[-1]
optimal_cost = gd_costs[-1]

# =============================================================================
# بهینه‌سازی چندهدفه با الگوریتم ژنتیک
# =============================================================================
class GasPipelineMultiObjective(Problem):
    def __init__(self):
        """
        بهینه‌سازی چندهدفه خط لوله گاز
        متغیرهای تصمیم:
          D: قطر لوله (اینچ)
          L: طول لوله (متر)

        اهداف:
          1. حداقل‌سازی هزینه کل
          2. حداقل‌سازی تلفات انرژی
          3. حداکثرسازی ظرفیت انتقال
        """
        super().__init__(n_var=2, n_obj=3, n_constr=0,
                         xl=np.array([5, 500]),
                         xu=np.array([50, 5000]))

    def _evaluate(self, X, out, *args, **kwargs):
        D = X[:, 0]
        L = X[:, 1]

        # تابع هدف 1: هزینه کل
        material_cost = 0.02 * D**2 * L
        installation_cost = 50 * L
        energy_cost = 10000 / D**4
        f1 = material_cost + installation_cost + energy_cost

        # تابع هدف 2: تلفات انرژی (کمینه‌سازی)
        f2 = 1000 / D**4  # تلفات انرژی متناسب با معکوس توان چهارم قطر

        # تابع هدف 3: ظرفیت انتقال (ماکزیمم‌سازی)
        # ظرفیت انتقال متناسب با مساحت مقطع و فشار (فرضی)
        capacity = 0.8 * D**2 * np.sqrt(L)
        f3 = -capacity  # منفی برای مینیمم‌سازی

        out["F"] = np.column_stack([f1, f2, f3])

# ایجاد و حل مسئله چندهدفه
problem = GasPipelineMultiObjective()
algorithm = NSGA2(pop_size=100)
res = minimize_moo(problem, algorithm, ('n_gen', 50), seed=1, verbose=False)

# استخراج نتایج
solutions = res.X
objectives = res.F

# جداسازی مقادیر
diameters = solutions[:, 0]
lengths = solutions[:, 1]
costs = objectives[:, 0]
energy_losses = objectives[:, 1]
capacities = -objectives[:, 2]  # تبدیل به مثبت برای ظرفیت واقعی

# =============================================================================
# ترسیم نمودارهای مقایسه‌ای
# =============================================================================
plt.figure(figsize=(18, 12))
plt.suptitle(persian_text('مقایسه روش‌های بهینه‌سازی در طراحی خط لوله گاز'),
             fontsize=22, y=0.98)

# ساختار GridSpec برای چیدمان نمودارها
gs = plt.GridSpec(2, 2, width_ratios=[1, 1.2], height_ratios=[1, 1])

# نمودار 1: همگرایی الگوریتم‌های تک‌هدفه
ax1 = plt.subplot(gs[0])
ax1.plot(gd_costs, 'ro-', alpha=0.7, label=persian_text('گرادیان کاهشی'))
ax1.plot(newton_costs, 'bs-', alpha=0.7, label=persian_text('روش نیوتن'))
ax1.axhline(optimal_cost, color='g', linestyle='--',
           label=persian_text(f'هزینه بهینه: {optimal_cost:.2f} دلار'))
ax1.set_xlabel(persian_text('تعداد تکرار'), fontsize=12)
ax1.set_ylabel(persian_text('هزینه کل (دلار)'), fontsize=12)
ax1.set_title(persian_text('همگرایی الگوریتم‌های بهینه‌سازی تک‌هدفه'), fontsize=14)
ax1.legend()
ax1.grid(alpha=0.3)

# نمودار 2: مسیرهای بهینه‌سازی در فضای طراحی
ax2 = plt.subplot(gs[1])
D_vals = np.linspace(5, 50, 100)
costs_vals = [gas_pipeline_cost(D, L_fixed) for D in D_vals]

ax2.plot(D_vals, costs_vals, 'b-', label=persian_text('تابع هزینه'))
ax2.plot(gd_path, gd_costs, 'ro-', alpha=0.7,
        label=persian_text('گرادیان کاهشی'))
ax2.plot(newton_path, newton_costs, 'gs-', alpha=0.7,
        label=persian_text('روش نیوتن'))
ax2.axvline(optimal_D, color='k', linestyle='--',
           label=persian_text(f'قطر بهینه: {optimal_D:.2f} اینچ'))
ax2.set_xlabel(persian_text('قطر لوله (اینچ)'), fontsize=12)
ax2.set_ylabel(persian_text('هزینه کل (دلار)'), fontsize=12)
ax2.set_title(persian_text('مسیرهای بهینه‌سازی در فضای طراحی'), fontsize=14)
ax2.legend()
ax2.grid(alpha=0.3)

# نمودار 3: جبهه پارتو سه‌بعدی (الگوریتم ژنتیک)
ax3 = plt.subplot(gs[2], projection='3d')
sc = ax3.scatter(costs, energy_losses, capacities,
                c=lengths, cmap='viridis', s=50, alpha=0.8)

# جهت‌های بهینه‌سازی
ax3.quiver(0, 0, 0, 100000, 0, 0, color='r', arrow_length_ratio=0.1,
          label=persian_text('کاهش هزینه'))
ax3.quiver(0, 0, 0, 0, 0.1, 0, color='g', arrow_length_ratio=0.1,
          label=persian_text('کاهش تلفات'))
ax3.quiver(0, 0, 0, 0, 0, 50000, color='b', arrow_length_ratio=0.1,
          label=persian_text('افزایش ظرفیت'))

ax3.set_xlabel(persian_text('هزینه کل'), fontsize=10, labelpad=15)
ax3.set_ylabel(persian_text('تلفات انرژی'), fontsize=10, labelpad=15)
ax3.set_zlabel(persian_text('ظرفیت انتقال'), fontsize=10, labelpad=15)
ax3.set_title(persian_text('جبهه پارتو سه‌بعدی (الگوریتم ژنتیک)'), fontsize=14)
ax3.legend()
ax3.view_init(elev=25, azim=45)

cbar = plt.colorbar(sc, ax=ax3, pad=0.1)
cbar.set_label(persian_text('طول لوله (متر)'), fontsize=10)

# نمودار 4: مقایسه راه‌حل‌های بهینه
ax4 = plt.subplot(gs[3])

# راه‌حل‌های مختلف
solutions = [
    {'name': persian_text('گرادیان کاهشی'),
     'D': optimal_D,
     'L': L_fixed,
     'cost': optimal_cost,
     'color': 'red'},

    {'name': persian_text('روش نیوتن'),
     'D': newton_path[-1],
     'L': L_fixed,
     'cost': newton_costs[-1],
     'color': 'green'},

    {'name': persian_text('ژنتیک-کم هزینه'),
     'D': diameters[np.argmin(costs)],
     'L': lengths[np.argmin(costs)],
     'cost': np.min(costs),
     'color': 'blue'},

    {'name': persian_text('ژنتیک-ظرفیت بالا'),
     'D': diameters[np.argmax(capacities)],
     'L': lengths[np.argmax(capacities)],
     'cost': costs[np.argmax(capacities)],
     'color': 'purple'}
]

# رسم مقایسه‌ای
width = 0.2
indices = np.arange(len(solutions))

# قطر لوله
ax4.bar(indices - width, [s['D'] for s in solutions], width,
       color=[s['color'] for s in solutions],
       label=persian_text('قطر لوله (اینچ)'))

# طول لوله
ax4.bar(indices, [s['L'] for s in solutions], width,
       color=[s['color'] for s in solutions],
       label=persian_text('طول لوله (متر)'))

# هزینه (در مقیاس ثانویه)
ax4_twin = ax4.twinx()
ax4_twin.plot(indices, [s['cost'] for s in solutions], 'ko-',
             markersize=8, label=persian_text('هزینه کل (دلار)'))

# تنظیمات محور
ax4.set_xticks(indices)
ax4.set_xticklabels([s['name'] for s in solutions], fontsize=12)
ax4.set_ylabel(persian_text('اندازه (اینچ/متر)'), fontsize=12)
ax4_twin.set_ylabel(persian_text('هزینه کل (دلار)'), fontsize=12)
ax4.set_title(persian_text('مقایسه راه‌حل‌های بهینه'), fontsize=14)

# جمع کردن راهنماها
lines, labels = ax4.get_legend_handles_labels()
lines2, labels2 = ax4_twin.get_legend_handles_labels()
ax4.legend(lines + lines2, labels + labels2, loc='upper left')

# خطوط راهنما برای هزینه
for i, s in enumerate(solutions):
    ax4_twin.text(i, s['cost'], f"{s['cost']:,.0f}",
                 ha='center', va='bottom', fontsize=10)

# تنظیم فاصله‌ها
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.subplots_adjust(wspace=0.25, hspace=0.3)

plt.savefig('gas_pipeline_optimization_comparison.png', dpi=300, bbox_inches='tight')
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from arabic_reshaper import reshape
from bidi.algorithm import get_display

# تابع برای نمایش متن فارسی
def persian_text(text):
    reshaped = reshape(text)
    return get_display(reshaped)

# تعریف تابع هزینه (تابع هدف)
def cost_function(x, y):
    term1 = 0.01 * (x**2 + y**2)
    term2 = 5 * np.exp(-((x - 3)**2 + (y - 2)**2))
    term3 = 2 * np.exp(-((x + 4)**2 + (y + 5)**2))
    return term1 - term2 - term3

# گرادیان تابع هدف
def gradient(x, y):
    dfdx = 0.02 * x + 2 * (x - 3) * 5 * np.exp(-((x - 3)**2 + (y - 2)**2)) + 2 * (x + 4) * 2 * np.exp(-((x + 4)**2 + (y + 5)**2))
    dfdy = 0.02 * y + 2 * (y - 2) * 5 * np.exp(-((x - 3)**2 + (y - 2)**2)) + 2 * (y + 5) * 2 * np.exp(-((x + 4)**2 + (y + 5)**2))
    return np.array([dfdx, dfdy])

# هسین تابع هدف (برای روش نیوتن)
def hessian(x, y):
    eps = 1e-5
    grad_x1 = gradient(x + eps, y)
    grad_x2 = gradient(x - eps, y)
    d2fdx2 = (grad_x1[0] - grad_x2[0]) / (2 * eps)

    grad_y1 = gradient(x, y + eps)
    grad_y2 = gradient(x, y - eps)
    d2fdy2 = (grad_y1[1] - grad_y2[1]) / (2 * eps)

    return np.array([[d2fdx2, 0],
                     [0, d2fdy2]])

# روش گرادیان کاهشی
def gradient_descent(start, lr=0.1, steps=30):
    path = [start]
    x, y = start
    for _ in range(steps):
        grad = gradient(x, y)
        x, y = x - lr * grad
        path.append((x, y))
    return np.array(path)

# روش نیوتن
def newton_method(start, steps=10):
    path = [start]
    x, y = start
    for _ in range(steps):
        grad = gradient(x, y)
        H = hessian(x, y)
        # پرش نیوتنی (فرض استقلال مشتقات)
        x = x - grad[0] / H[0, 0] if H[0, 0] != 0 else x
        y = y - grad[1] / H[1, 1] if H[1, 1] != 0 else y
        path.append((x, y))
    return np.array(path)

# مش‌بندی برای رسم سطح تابع
x_vals = np.linspace(-10, 10, 100)
y_vals = np.linspace(-10, 10, 100)
X, Y = np.meshgrid(x_vals, y_vals)
Z = cost_function(X, Y)

# نقطه شروع مشترک
start_point = np.array([-8.0, 8.0])
gd_path = gradient_descent(start_point, lr=0.1, steps=30)
newton_path = newton_method(start_point, steps=10)

# رسم نمودار ۳بعدی
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.7)

# مسیرها
z_gd = cost_function(gd_path[:,0], gd_path[:,1])
ax.plot(gd_path[:,0], gd_path[:,1], z_gd, marker='o', linestyle='--', label=persian_text('گرادیان کاهشی'))

z_newton = cost_function(newton_path[:,0], newton_path[:,1])
ax.plot(newton_path[:,0], newton_path[:,1], z_newton, marker='^', linestyle='-.', label=persian_text('روش نیوتن'))

# برچسب‌ها
ax.set_title(persian_text('مسیر بهینه‌سازی دو الگوریتم روی تابع هزینه حفاری'), fontsize=14)
ax.set_xlabel(persian_text('x: موقعیت شرق-غرب'), fontsize=12)
ax.set_ylabel(persian_text('y: موقعیت شمال-جنوب'), fontsize=12)
ax.set_zlabel(persian_text('هزینه حفاری'), fontsize=12)
ax.legend()

plt.tight_layout()
plt.show()
````

````python
# Re-run after code execution state reset
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from arabic_reshaper import reshape
from bidi.algorithm import get_display

# تابع برای نمایش متن فارسی
def persian_text(text):
    reshaped = reshape(text)
    return get_display(reshaped)

# مرحله ۱: داده‌های فرضی واقعی (x, y, cost)
np.random.seed(0)
x_data = np.random.uniform(-5, 5, 50)
y_data = np.random.uniform(-5, 5, 50)
# فرض تابع هزینه واقعی: ترکیب چندجمله‌ای درجه 2 + نویز
cost_data = 3 + 2*x_data + y_data**2 - 0.5*x_data*y_data + np.random.normal(0, 2, 50)

# مرحله ۲: برازش مدل رگرسیون چندجمله‌ای درجه 2
X_train = np.vstack([x_data, y_data]).T
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X_train)
model = LinearRegression().fit(X_poly, cost_data)

# مرحله ۳: ساخت مش برای ترسیم سطح
x_range = np.linspace(-6, 6, 100)
y_range = np.linspace(-6, 6, 100)
X_mesh, Y_mesh = np.meshgrid(x_range, y_range)
XY_mesh = np.c_[X_mesh.ravel(), Y_mesh.ravel()]
Z_mesh = model.predict(poly.transform(XY_mesh)).reshape(X_mesh.shape)

# مرحله ۴: تعریف تابع هزینه و گرادیان‌ها برای بهینه‌سازی
def cost_function(x, y):
    input_poly = poly.transform([[x, y]])
    return model.predict(input_poly)[0]

def gradient(x, y):
    eps = 1e-4
    dx = (cost_function(x + eps, y) - cost_function(x - eps, y)) / (2 * eps)
    dy = (cost_function(x, y + eps) - cost_function(x, y - eps)) / (2 * eps)
    return np.array([dx, dy])

def hessian(x, y):
    eps = 1e-4
    dxx = (cost_function(x + eps, y) - 2*cost_function(x, y) + cost_function(x - eps, y)) / (eps**2)
    dyy = (cost_function(x, y + eps) - 2*cost_function(x, y) + cost_function(x, y - eps)) / (eps**2)
    dxy = (cost_function(x + eps, y + eps) - cost_function(x + eps, y - eps) -
           cost_function(x - eps, y + eps) + cost_function(x - eps, y - eps)) / (4 * eps**2)
    return np.array([[dxx, dxy], [dxy, dyy]])

# مرحله ۵: گرادیان کاهشی
path_gd = []
x, y = -4.5, -4.5
for _ in range(30):
    path_gd.append((x, y, cost_function(x, y)))
    grad = gradient(x, y)
    x, y = x - 0.1 * grad[0], y - 0.1 * grad[1]

# مرحله ۶: روش نیوتن
path_nt = []
x, y = -4.5, -4.5
for _ in range(10):
    path_nt.append((x, y, cost_function(x, y)))
    grad = gradient(x, y)
    hess = hessian(x, y)
    try:
        delta = np.linalg.solve(hess, grad)
        x, y = x - delta[0], y - delta[1]
    except np.linalg.LinAlgError:
        break  # اگر هسین منفرد بود، ادامه نده

# تبدیل مسیرها به آرایه برای ترسیم
path_gd = np.array(path_gd)
path_nt = np.array(path_nt)

# مرحله ۷: ترسیم نهایی
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# سطح هزینه
ax.plot_surface(X_mesh, Y_mesh, Z_mesh, cmap='viridis', alpha=0.7)

# نقاط داده اصلی
ax.scatter(x_data, y_data, cost_data, c='r', marker='x', label=persian_text('داده‌ها'))

# مسیر گرادیان کاهشی
ax.plot(path_gd[:,0], path_gd[:,1], path_gd[:,2], color='blue', marker='o', label=persian_text('گرادیان کاهشی'))

# مسیر نیوتن
ax.plot(path_nt[:,0], path_nt[:,1], path_nt[:,2], color='orange', marker='^', label=persian_text('روش نیوتن'))

ax.set_xlabel("x (km)")
ax.set_ylabel("y (km)")
ax.set_zlabel(persian_text("هزینه حفاری (میلیون دلار)"))
#ax.set_title(persian_text("بهینه‌سازی محل حفاری "))
ax.legend()

plt.tight_layout()
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from pymoo.core.problem import Problem
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.termination import get_termination
from pymoo.optimize import minimize
from pymoo.core.callback import Callback
from pymoo.visualization.scatter import Scatter
from arabic_reshaper import reshape
from bidi.algorithm import get_display

# تابع برای نمایش متن فارسی
def persian_text(text):
    reshaped = reshape(text)
    return get_display(reshaped)

# =============================================================================
# تعریف تابع هدف بر اساس داده‌های واقعی (cost و -production)
# =============================================================================
class DrillingProblem(Problem):
    def __init__(self, cost_fun, prod_fun):
        super().__init__(n_var=2, n_obj=2, n_constr=0, xl=-5, xu=5)
        self.cost_fun = cost_fun
        self.prod_fun = prod_fun

    def _evaluate(self, X, out, *args, **kwargs):
        cost = self.cost_fun(X)
        production = self.prod_fun(X)
        out["F"] = np.column_stack([cost, -production])  # minimize both

# =============================================================================
# توابع هزینه و برداشت برحسب x و y
# =============================================================================
def cost_function(X):
    x, y = X[:,0], X[:,1]
    noise = np.random.normal(0, 0.2, len(x))
    return 10 + 0.5*x**2 + 0.3*y**2 + noise

def prod_function(X):
    x, y = X[:,0], X[:,1]
    return 5 + 2*np.exp(-0.1*(x-1)**2 - 0.2*(y+2)**2) + np.random.normal(0, 0.1, len(x))

problem = DrillingProblem(cost_function, prod_function)

# =============================================================================
# ثبت مسیر تکامل جمعیت
# =============================================================================
class MyCallback(Callback):
    def __init__(self):
        super().__init__()
        self.data["pop"] = []

    def notify(self, algorithm):
        self.data["pop"].append(algorithm.pop.get("F"))

callback = MyCallback()

# =============================================================================
# اجرای NSGA-II
# =============================================================================
algorithm = NSGA2(pop_size=100)
res = minimize(problem,
               algorithm,
               termination=get_termination("n_gen", 40),
               callback=callback,
               seed=1,
               verbose=False)

# =============================================================================
# ترسیم مسیر تکامل جمعیت
# =============================================================================
colors = ['gray', 'blue', 'green', 'red']
gens = [0, 10, 20, -1]  # نسل‌های انتخابی: اولیه، میانه، نهایی

plt.figure(figsize=(8,6))
for i, gen in enumerate(gens):
    pop = callback.data["pop"][gen]
    plt.scatter(pop[:,0], pop[:,1],
                label=persian_text(f"نسل {gen if gen != -1 else len(callback.data['pop'])-1}"),
                alpha=0.6, s=50, color=colors[i], edgecolor='k')

plt.xlabel(persian_text("هزینه حفاری"))
plt.ylabel(persian_text("منفی برداشت"))
plt.title(persian_text("مسیر تکامل جمعیت NSGA-II در بهینه‌سازی چندهدفه"))
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
````

````python
بر اساس توضیحات شما، دیتاست `Groningen_test.csv` شامل ویژگی‌های زیر است:

### **۱. شناسه‌ها و متادیتا**
- **WELLNAME**: نام چاه (شناسه سری زمانی)
- **DEPTH**: عمق اندازه‌گیری (بر حسب متر یا فوت)

### **۲. ویژگی‌های پتروفیزیکی (اندازه‌گیری‌های لاگینگ)**
| نام ستون | توضیحات | واحد اندازه‌گیری |
|----------|---------|------------------|
| **DENS** | چگالی سنجی اول | g/cm³ |
| **DEN** | چگالی سنجی دوم | g/cm³ |
| **DT** | سونیک اول (زمان عبور موج صوتی) | μs/ft |
| **SONIC** | سونیک دوم | μs/ft |
| **FLGR** | پرتوزایی گاما اول | gAPI |
| **GR** | پرتوزایی گاما دوم | gAPI |
| **CAL** | اندازه‌گیری کالیپر (قطر چاه) | اینچ |
| **RED** | مقاومت الکتریکی اول | Ω·m |
| **RES_DEP** | مقاومت الکتریکی دوم (احتمالاً عمقی) | Ω·m |

### **۳. برچسب هدف**
- **target**: نشانگر نواحی خاص در چاه (مثلاً لایه‌های دارای هیدروکربن، مناطق مشکل‌ساز)

---

### **تفسیر تخصصی ویژگی‌ها**
این دیتاست یک مجموعه داده **چندسنجشی (Multi-Measurement)** از چاه‌های نفتی است که شامل:

1. **لاگ‌های فیزیکی:**
   - چگالی (`DENS`, `DEN`): برای شناسایی تخلخل سنگ
   - سونیک (`DT`, `SONIC`): اندازه‌گیری سرعت صوت برای ارزیابی الاستیسیته سنگ
   - گاما (`FLGR`, `GR`): تشخیص سنگ‌های رسی (شیل)
   - مقاومت (`RED`, `RES_DEP`): شناسایی سیالات درون سازند

2. **اندازه‌های عملیاتی:**
   - کالیپر (`CAL`): نشان‌دهنده پایداری دیواره چاه
     - مقادیر > قطر اولیه: نشانه ریزش دیواره
     - مقادیر < قطر اولیه: نشانه تجمع رسوبات

3. **نشانگرهای زمین‌شناسی:**
   - ترکیب `GR` و `RES_DEP`: تشخیص لیتولوژی
   - نسبت `DT` به `DENS`: شناسایی مناطق شکستگی

---

### **کاربردها در بهینه‌سازی حفاری**
با این داده‌ها می‌توان:
```python
# مثال: تشخیص نواحی پرخطر بر اساس ترکیب لاگ‌ها
df['RISK_ZONE'] = np.where(
    (df['CAL'] > 12.5) |  # ریزش دیواره
    (df['GR'] > 120) |    # شیل زیاد
    (df['RES_DEP'] < 5),  # آب زیرزمینی
    1, 0  # پرخطر/بی‌خطر
)
```

### **روابط کلیدی برای بهینه‌سازی**
1. **زمان حفاری ∝ عمق × سختی سنگ**
   `HARDNESS = (GR / RES_DEP) * DENS`

2. **هزینه ∝ انرژی مصرفی + افت تجهیزات**
   `COST = f(CAL, GR)` (نوسانات کالیپر و گاما)

3. **ریسک ایمنی ∝ احتمال ریزش + فشار غیرعادی**
   `RISK = g(DT, CAL.diff())` (تغییرات ناگهانی در سونیک و کالیپر)

این ویژگی‌ها به‌طور مستقیم در توابع هدف بهینه‌سازی چندهدفه که پیش‌تر پیاده‌سازی شد استفاده می‌شوند.
````

````python
from pymoo.core.problem import Problem
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize
from pymoo.visualization.scatter import Scatter

class DrillingProblem(Problem):
    def __init__(self):
        super().__init__(
            n_var=len(features),
            n_obj=3,
            xl=np.array([df[f].min() for f in features]),
            xu=np.array([df[f].max() for f in features])
        )

    def _evaluate(self, X, out, *args, **kwargs):
        F = np.array([objective_function(x) for x in X])
        out["F"] = F

problem = DrillingProblem()

algorithm = NSGA2(pop_size=100)
res = minimize(
    problem,
    algorithm,
    ('n_gen', 200),
    seed=42,
    verbose=True
)

# ذخیره نتایج
pareto_front = res.F
````

````python
from statsmodels.tsa.stattools import adfuller

# آزمون ADF برای ایستایی
result = adfuller(df['Price'].dropna())
print(f"ADF Statistic: {result[0]}")
print(f"p-value: {result[1]}")

# تفاضل‌گیری در صورت لزوم
if result[1] > 0.05:
    df['Price_diff'] = df['Price'].diff()
    df.dropna(inplace=True)
else:
    df['Price_diff'] = df['Price']
````

````python
مرحله 3: مدل‌سازی با ARIMA
````

````python
from statsmodels.tsa.arima.model import ARIMA

model_arima = ARIMA(df['Price'], order=(1,1,1))
model_arima_fit = model_arima.fit()

# خلاصه مدل
print(model_arima_fit.summary())

# پیش‌بینی
forecast_arima = model_arima_fit.forecast(steps=30)
forecast_arima.plot(title='پیش‌بینی 30 روزه با ARIMA')
plt.show()
````

````python
مرحله 4: مدل XGBoost
````

````python
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np

# ساخت ویژگی‌های lag برای مدل XGBoost
for i in range(1, 8):
    df[f'lag_{i}'] = df['Price'].shift(i)

df.dropna(inplace=True)

# تقسیم داده
train = df.iloc[:-30]
test = df.iloc[-30:]

X_train = train[[f'lag_{i}' for i in range(1, 8)]]
y_train = train['Price']

X_test = test[[f'lag_{i}' for i in range(1, 8)]]
y_test = test['Price']

# آموزش مدل XGBoost
model_xgb = XGBRegressor()
model_xgb.fit(X_train, y_train)

# پیش‌بینی
pred_xgb = model_xgb.predict(X_test)

# ارزیابی
rmse_xgb = np.sqrt(mean_squared_error(y_test, pred_xgb))
mae_xgb = mean_absolute_error(y_test, pred_xgb)
print(f"XGBoost → RMSE: {rmse_xgb:.2f}, MAE: {mae_xgb:.2f}")

# رسم
plt.figure(figsize=(10,5))
plt.plot(y_test.index, y_test.values, label='واقعی')
plt.plot(y_test.index, pred_xgb, label='پیش‌بینی XGBoost')
plt.legend()
plt.title("پیش‌بینی XGBoost")
plt.show()
````

````python
مرحله 5: مدل LSTM با Keras
````

````python
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM

# نرمال‌سازی
scaler = MinMaxScaler()
scaled = scaler.fit_transform(df[['Price']])

# ساخت دیتاست با window
def create_dataset(data, window_size=7):
    X, y = [], []
    for i in range(len(data) - window_size):
        X.append(data[i:i+window_size])
        y.append(data[i+window_size])
    return np.array(X), np.array(y)

X_lstm, y_lstm = create_dataset(scaled)

# تقسیم train/test
split = len(X_lstm) - 30
X_train_lstm, X_test_lstm = X_lstm[:split], X_lstm[split:]
y_train_lstm, y_test_lstm = y_lstm[:split], y_lstm[split:]

# بازشکل‌دهی
X_train_lstm = X_train_lstm.reshape((X_train_lstm.shape[0], X_train_lstm.shape[1], 1))
X_test_lstm = X_test_lstm.reshape((X_test_lstm.shape[0], X_test_lstm.shape[1], 1))

# ساخت مدل
model_lstm = Sequential()
model_lstm.add(LSTM(50, activation='relu', input_shape=(X_train_lstm.shape[1], 1)))
model_lstm.add(Dense(1))
model_lstm.compile(optimizer='adam', loss='mse')

# آموزش
model_lstm.fit(X_train_lstm, y_train_lstm, epochs=20, verbose=0)

# پیش‌بینی و بازگردانی مقیاس
pred_lstm = model_lstm.predict(X_test_lstm)
pred_lstm = scaler.inverse_transform(pred_lstm)
y_true_lstm = scaler.inverse_transform(y_test_lstm.reshape(-1,1))

# ارزیابی
rmse_lstm = np.sqrt(mean_squared_error(y_true_lstm, pred_lstm))
mae_lstm = mean_absolute_error(y_true_lstm, pred_lstm)
print(f"LSTM → RMSE: {rmse_lstm:.2f}, MAE: {mae_lstm:.2f}")

# رسم
plt.plot(y_true_lstm, label='واقعی')
plt.plot(pred_lstm, label='پیش‌بینی LSTM')
plt.legend()
plt.title("پیش‌بینی LSTM")
plt.show()
````

````python

✅ ۱. مدل‌سازی پویا یعنی چه؟

در اینجا "مدل‌سازی پویا" یعنی:

> پیش‌بینی قیمت آینده نفت براساس داده‌های گذشته (سری زمانی) با درنظر گرفتن تغییرات در زمان.




---

📉 آیا قیمت نفت قابلیت پیش‌بینی دارد؟

بله ولی نه دقیق و همیشه. قیمت نفت:

بخشی الگوی فصلی دارد (مثلاً مصرف زمستانی/تابستانی).

تا حدی هم وابسته به روند گذشته است.

اما بخش بزرگی از آن تصادفی و متاثر از رویدادهای سیاسی/اقتصادی جهانی است.



---

🔍 چرا پیش‌بینی بلندمدت سخت است؟

چون:

تصمیم‌های اوپک، تحریم‌ها، جنگ‌ها، فناوری‌ها، کرونا، و حتی تغییر ارز دلار می‌توانند باعث شوک قیمتی شوند.

هیچ‌کدام از این‌ها در سری زمانی گذشته وجود ندارند و بنابراین:

> «مدل نمی‌تواند چیزی را پیش‌بینی کند که در تاریخش ندیده.»





---

🧠 مدل‌ها چه چیزهایی را خوب پیش‌بینی می‌کنند؟

مدل مزایا   محدودیت‌ها

ARIMA   مناسب برای روندهای خطی، کوتاه‌مدت   دقت پایین در نوسانات شدید
XGBoost قدرت در یادگیری روابط پیچیده بین lagها  به رویدادهای ناگهانی حساس نیست
LSTM    خوب برای الگوهای غیرخطی و حافظه‌دار نیاز به داده زیاد، overfit روی نویز



---

🎯 چقدر قابل اعتماد هستند؟

🔹 پیش‌بینی کوتاه‌مدت (۵ تا ۱۵ روز آینده):

نسبتاً قابل اعتماد با خطای حدود ۵٪ تا ۱۰٪ (اگر داده تمیز و خوب باشد)

مخصوصاً اگر نوسانات شدید یا بحران خاصی وجود نداشته باشد.


🔹 پیش‌بینی میان‌مدت (۱ تا ۳ ماه آینده):

دقت کاهش می‌یابد، خطا تا ۱۵٪ یا بیشتر ممکن است.


🔹 پیش‌بینی بلندمدت (بیش از ۶ ماه):

غیرقابل اعتماد با روش‌های سری زمانی تنها.

برای این‌کار نیاز به مدل‌های ترکیبی داریم که شامل:

داده‌های اقتصادی (GDP، عرضه/تقاضا جهانی)

شاخص دلار، نرخ بهره

رویدادهای ژئوپلتیک

NLP روی اخبار و داده‌های متنی




---

📌 جمع‌بندی:

طول پیش‌بینی    دقت قابل اتکا برای تصمیم‌گیری؟

کوتاه‌مدت (تا 15 روز)   ✅ خوب   ✅ بله
میان‌مدت (1-3 ماه)  ⚠️ متوسط    ⚠️ با احتیاط
بلندمدت (بیش از 3 ماه)  ❌ ضعیف  ❌ نه، مگر با مدل‌های پیچیده‌تر



---

اگر بخوای مدل‌های ترکیبی‌تر با تحلیل اخبار یا شرایط اقتصاد کلان بسازیم، می‌تونم راهنمایی کامل‌تری هم بدم.
همچنین اگر هدفت پیش‌بینی برای تصمیم‌گیری سرمایه‌گذاری یا صنعتی باشه، می‌تونیم روش‌های مقاوم‌تر مثل مدل‌های Bayesian، Prophet+Exogenous، یا Multi-modal پیشنهاد بدیم.

آیا هدف خاصی برای استفاده از این پیش‌بینی‌ها داری؟ مثلاً در سرمایه‌گذاری، برنامه‌ریزی تولید یا تجارت؟
````

````python
مدلسازی تصادفی (Stochastic Modeling) در حوزه نفت و گاز ابزاری حیاتی برای مدیریت **عدمقطعیتها** در دادههای پیچیده زمینشناسی، مخازن و فرآیندهای تولید است. این رویکرد به جای ارائه یک پاسخ قطعی، طیفی از نتایج محتمل را ارائه میدهد که تصمیمگیری در شرایط ریسک را بهبود میبخشد. در ادامه، کاربردهای کلیدی آن در این صنعت را بررسی میکنیم:

---

### **۱. دادهیابی (Data Acquisition) و پردازش داده‌ها**
- **طراحی بهینه شبکههای لرزه‌نگاری و چاه‌های اکتشافی**:
  با استفاده از روش‌های تصادفی (مانند شبیهسازی مونت کارلو)، موقعیت چاه‌ها و حسگرها بهینه‌سازی میشود تا عدمقطعیت مدل زمینشناسی کاهش یابد.
- **پردازش داده‌های لرزه‌ای**:
  مدل‌های تصادفی برای فیلترکردن نویز، بازسازی داده‌های مفقود شده و بهبود کیفیت تصاویر زیرسطحی به کار میروند.

---

### **۲. مدلسازی مخزن (Reservoir Modeling)**
- **شبیهسازی تصادفی مخازن (Stochastic Reservoir Simulation)**:
  - تولید **چندین سناریوی محتمل** از توزیع خصوصیات مخزن (تراوایی، تخلخل، اشباع سیالات) با استفاده از روش‌هایی مانند:
    - **Kriging** و **Sequential Gaussian Simulation (SGS)** برای متغیرهای پیوسته.
    - **Sequential Indicator Simulation (SIS)** برای متغیرهای رسته‌ای (مانند واحدهای سنگی).
  - **مدلسازی هندسه لایه‌ها و گسل‌ها** با تکنیک‌های **Object-Based Modeling**.
- **ارزیابی عدمقطعیت**:
  خروجی مدل‌ها به صورت **توزیع‌های احتمال** (مثلاً توزیع نفت درجا، تولید تجمعی) ارائه میشود.

---

### **۳. ارزیابی مخزن و مدیریت تولید (Reservoir Evaluation & Production Optimization)**
- **پیش‌بینی تولید با روش‌های تصادفی**:
  - ترکیب مدل‌های مخزن تصادفی با **شبیهسازهای عددی** (مثل Eclipse یا CMG) برای پیش‌بینی دبی تولید تحت شرایط مختلف.
  - آنالیز **حساسیت** و **ریسک** پارامترهای نامطمئن (مثل فشار مخزن، رفتار چاه‌ها).
- **بهینه‌سازی توسعه مخزن**:
  تعیین مکان بهینه چاه‌های جدید، استراتژیهای تزریق آب/گاز و زمانبندی تولید بر اساس تحلیل عدمقطعیت.

---

### **۴. ارزیابی اقتصادی پروژه‌ها (Economic Evaluation)**
- **شبیهسازی مونت کارلو برای NPV و ROI**:
  ترکیب عدمقطعیت‌های فنی (مثلاً حجم مخزن، بازیافت نهایی) با عوامل اقتصادی (قیمت نفت، هزینه‌ها) برای محاسبه توزیع احتمال شاخص‌های اقتصادی.
- **تخصیص سرمایه و مدیریت ریسک**:
  اولویت‌بندی پروژه‌ها بر اساس **منحنیهای توزیع NPV** و احتمال دستیابی به سودآوری.

---

### **۵. پیش‌بینی خرابی تجهیزات (Predictive Maintenance)**
- **مدل‌های تصادفی برای آنالیز قابلیت اطمینان**:
  پیش‌بینی زمان خرابی تجهیزات حیاتی (مثل پمپ‌ها، شیرها) با استفاده از داده‌های حسگرها و روش‌هایی مانند **زنجیره مارکوف**.

---

### **۶. چالش‌های کلیدی**
- **حجم بالای داده‌ها و محاسبات**: نیاز به ابررایانه‌ها و الگوریتم‌های کارآمد (مثل تقریب‌های آماری).
- **ادغام داده‌های ناهمگن**: تلفیق داده‌های لرزه‌ای، چاه‌ها، آزمایش‌های مغزه و تولید.
- **ارزیابی کیفیت مدل**: سنجش دقت پیش‌بینی‌ها با معیارهایی مانند **تاریخچه تطبیق تولید (History Matching)**.

---

### **ابزارهای رایج**
- **زبان‌های برنامه‌نویسی**: Python (با کتابخانه‌های `NumPy`, `SciPy`, `PyMC3`)، R.
- **نرم‌افزارهای تخصصی**: Petrel (با ماژول Uncertainty Analysis)، CMG، GSLIB، TensorFlow برای مدل‌های مبتنی بر یادگیری ماشین.

---

### **جمع‌بندی**
مدلسازی تصادفی در نفت و گاز، **پل ارتباطی** بین داده‌های پراکنده و تصمیم‌های استراتژیک است. با کمّی‌سازی عدمقطعیت‌ها، به مدیران اجازه میدهد:
- پروژه‌های پرریسک را شناسایی کنند،
- سناریوهای "بدترین حالت" را پیش‌بینی کنند،
- سرمایه‌گذاری‌ها را بر اساس احتمال موفقیت بهینه‌سازی نمایند.

آیا تمرکز خاصی بر یکی از بخش‌ها (مثل اکتشاف، تولید یا اقتصاد) دارید؟ میتوانم جزئیات فنی بیشتری ارائه دهم.
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# گام 1: خواندن داده‌ها
df = pd.read_csv('/mnt/data/4_5783066107061475916.csv')
df.columns = [c.strip().lower().replace(' ', '_') for c in df.columns]
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date').set_index('date')

# نمایش خلاصه‌ای از داده
print(df.head())

# گام 2: محاسبه بازده‌های روزانه (log returns)
df['log_return'] = np.log(df['price'] / df['price'].shift(1))
df = df.dropna()

# گام 3: تعریف حالت‌ها برای زنجیره مارکوف (گسسته‌سازی بازده)
num_states = 5
df['state'], bins = pd.qcut(df['log_return'], q=num_states, labels=False, retbins=True)

# گام 4: تخمین ماتریس انتقال مارکوف
transition_matrix = np.zeros((num_states, num_states))
for (i, j) in zip(df['state'][:-1], df['state'][1:]):
    transition_matrix[i, j] += 1
transition_matrix = transition_matrix / transition_matrix.sum(axis=1, keepdims=True)

# نمایش ماتریس انتقال
print("\nماتریس انتقال:")
print(np.round(transition_matrix, 3))

# گام 5: شبیه‌سازی مونت‌کارلو با استفاده از زنجیره مارکوف
def simulate_markov_mc(start_price, start_state, n_days, transition_matrix, bins, n_simulations=10):
    simulations = []
    for _ in range(n_simulations):
        prices = [start_price]
        state = start_state
        for _ in range(n_days):
            # انتخاب حالت بعدی بر اساس ماتریس انتقال
            state = np.random.choice(np.arange(num_states), p=transition_matrix[state])
            # انتخاب یک مقدار بازده تصادفی در بازه همان حالت
            ret = np.random.uniform(bins[state], bins[state + 1])
            price = prices[-1] * np.exp(ret)
            prices.append(price)
        simulations.append(prices)
    return np.array(simulations)

# پارامترهای اولیه
start_price = df['price'].iloc[-1]
start_state = df['state'].iloc[-1]
n_days = 100
n_simulations = 100

simulated_prices = simulate_markov_mc(start_price, start_state, n_days, transition_matrix, bins, n_simulations)

# گام 6: رسم نتایج شبیه‌سازی
plt.figure(figsize=(12, 6))
for i in range(n_simulations):
    plt.plot(simulated_prices[i], alpha=0.3)
plt.title('شبیه‌سازی مونت‌کارلو قیمت نفت با زنجیره مارکوف')
plt.xlabel('روزهای آینده')
plt.ylabel('قیمت شبیه‌سازی‌شده')
plt.grid(True)
plt.show()
````

````python
def simulate_well_paths(start_state, transition_matrix, bins,
                        start_prod, n_years=10, n_simulations=1000):
    """
    شبیه‌سازی n_simulations مسیر تولید برای یک چاه،
    با شبیه‌سازی وضعیت (state) و سپس تولید متناظر.
    """
    all_paths = np.zeros((n_simulations, n_years+1))
    all_paths[:, 0] = start_prod

    for sim in range(n_simulations):
        state = start_state
        prod = start_prod
        for t in range(1, n_years+1):
            # انتخاب state بعدی
            state = np.random.choice(n_states, p=transition_matrix[state])
            # نمونه‌گیری تولید در آن حالت به صورت یکنواخت در بین باین
            prod = np.random.uniform(bins[state], bins[state+1])
            all_paths[sim, t] = prod

    return all_paths

# 5. آرایه شبیه‌سازی برای یک چاه نمونه
# فرض کنیم چاهی با API مشخص و سال آخر داشتن یک state و MonthProd مشخص
sample_well = df['API_WellNo'].iloc[0]
last_record = df[df['API_WellNo']==sample_well].iloc[-1]
start_state = int(last_record['state'])
start_prod  = float(last_record['MonthProd'])

paths = simulate_well_paths(start_state, transition_matrix, bins,
                            start_prod, n_years=10, n_simulations=200)
````

````python
توضیحات نهایی

discretization
تولید هر چاه در هر سال به ۵ حالت (quintile) تقسیم شد.

transition_matrix
احتمال رفتن از هر حالت به حالت بعدی را از داده‌های واقعی (سال t → t+1) برآورد کردیم.

Monte Carlo
مسیرهای آینده را با ۱۰۰۰ شبیه‌سازی (یا کم‌تر/بیش‌تر بنا بر نیاز) تولید کرده و نتایج (میانه و بازه اطمینان) را رسم کردیم.
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# === ۱. بارگذاری و پیش‌پردازش عمومی ===
def load_and_prepare(path, filter_col=None, filter_vals=None):
    """
    path         : مسیر فایل CSV
    filter_col   : نام ستونی که می‌خواهی فیلتر کنی (مثلاً 'County')
    filter_vals  : لیستی از مقادیر مجاز برای آن ستون
    """
    df = pd.read_csv(path)
    # تنها ستون‌های لازم
    df = df[['API_WellNo', 'Year', 'MonthProd']].dropna()
    df['Year'] = df['Year'].astype(int)
    # اعمال فیلترِ دلخواه (اختیاری)
    if filter_col and filter_vals:
        df = df[df[filter_col].isin(filter_vals)]
    # مرتب‌سازی
    df = df.sort_values(['API_WellNo', 'Year'])
    return df

# === ۲. محاسبه ماتریس انتقال مارکوف ===
def compute_transition_matrix(df, n_states=5):
    """
    df      : دیتافریم آماده‌شده
    n_states: تعداد حالت‌ها (quintile, tercile و ...)
    بازمی‌گرداند: transition_matrix, bins
    """
    # گسسته‌سازی تولید به n_states
    df = df.copy()
    df['state'], bins = pd.qcut(df['MonthProd'], q=n_states, labels=False, retbins=True)
    # ساخت ماتریس شمارش
    P = np.zeros((n_states, n_states))
    for _, grp in df.groupby('API_WellNo'):
        s = grp['state'].values
        for (i, j) in zip(s[:-1], s[1:]):
            P[i, j] += 1
    # نرمال‌سازی
    with np.errstate(divide='ignore', invalid='ignore'):
        T = P / P.sum(axis=1, keepdims=True)
        T[np.isnan(T)] = 0
    return T, bins

# === ۳. شبیه‌سازی مونت‌کارلو برای یک چاه یا همه چاه‌ها ===
def simulate_paths(df, T, bins, n_states, n_years=10, n_sims=500, target_well=None):
    """
    df         : دیتافریم اصلی (عدم نیاز به state)
    T, bins    : از تابع compute_transition_matrix
    n_states   : تعداد حالت‌ها (به‌همراه T)
    n_years    : تا چه سالی شبیه‌سازی کند
    n_sims     : تعداد مسیرهای مونت‌کارلو
    target_well: اگر None باشد، همه چاه‌ها تجمیع؛
                 در غیر این صورت API_WellNo مشخص
    """
    # اگر برای یک چاه خاص:
    if target_well:
        last = df[df['API_WellNo']==target_well].iloc[-1]
        starts = [(int(last['state']), float(last['MonthProd']))]
    else:
        # تمام چاه‌ها: برای هر API یک شروع
        starts = []
        for well, grp in df.groupby('API_WellNo'):
            last = grp.iloc[-1]
            starts.append((int(last['state']), float(last['MonthProd'])))

    # شبیه‌سازی
    results = {}
    for idx, (s0, p0) in enumerate(starts):
        well_id = target_well or df['API_WellNo'].unique()[idx]
        paths = np.zeros((n_sims, n_years+1))
        paths[:, 0] = p0

        for sim in range(n_sims):
            state, prod = s0, p0
            for t in range(1, n_years+1):
                state = np.random.choice(n_states, p=T[state])
                prod = np.random.uniform(bins[state], bins[state+1])
                paths[sim, t] = prod

        results[well_id] = paths
    return results

# === ۴. رسم نتایج ===
def plot_simulation(results, kind='individual', ci=(5,95)):
    """
    results: دیکشنری {well_id: paths_array}
    kind   : 'individual' یا 'aggregate'
    ci     : درصد بازه اطمینان برای حالت تجمیعی
    """
    if kind == 'individual':
        for well, paths in results.items():
            plt.figure(figsize=(8,4))
            for row in paths:
                plt.plot(row, alpha=0.2)
            plt.title(f'مسیرهای تولید چاه {well}')
            plt.xlabel('سال‌های آینده')
            plt.ylabel('MonthProd')
            plt.grid(True)
            plt.show()
    else:  # aggregate: میانگین و بازه CI
        all_paths = np.vstack(list(results.values()))
        years = np.arange(all_paths.shape[1])
        median = np.median(all_paths, axis=0)
        lo = np.percentile(all_paths, ci[0], axis=0)
        hi = np.percentile(all_paths, ci[1], axis=0)

        plt.figure(figsize=(8,4))
        plt.plot(years, median, label='میانه')
        plt.fill_between(years, lo, hi, alpha=0.3, label=f'{ci[0]}–{ci[1]}٪')
        plt.title('مسیر تجمیعی تولید (میانه و بازه اطمینان)')
        plt.xlabel('سال‌های آینده')
        plt.ylabel('MonthProd')
        plt.legend()
        plt.grid(True)
        plt.show()

# === ====== مثال استفاده ====== ===
if __name__ == '__main__':
    # تنظیم پارامترها
    path = '/mnt/data/4_5783066107061475916.csv'
    df = load_and_prepare(path, filter_col='County', filter_vals=['Erie', 'Allegany'])
    T, bins = compute_transition_matrix(df, n_states=4)     # تغییر n_states
    sims = simulate_paths(df, T, bins, n_states=4,
                          n_years=8,    # تغییر افق شبیه‌سازی
                          n_sims=300,   # تغییر تعداد شبیه‌سازی
                          target_well=None)  # None → همه چاه‌ها
    # رسم تجمیعی
    plot_simulation(sims, kind='aggregate', ci=(10,90))
````

````python
# -*- coding: utf-8 -*-
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.inspection import PartialDependenceDisplay
import shap
from pygam import LinearGAM, s
from arabic_reshaper import reshape
from bidi.algorithm import get_display
import seaborn as sns

# --- تنظیمات نمایش فارسی ---
plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['axes.unicode_minus'] = False

def format_persian(text):
    """تابع تبدیل متن فارسی برای نمایش صحیح"""
    reshaped_text = reshape(text)
    return get_display(reshaped_text)

# --- خواندن و پیش‌پردازش داده‌ها ---
df = pd.read_excel('data.xls', parse_dates=['Date'], index_col='Date')

# نمایش ساختار داده‌ها
print("ستون‌های موجود در داده‌ها:")
print(df.columns.tolist())

# تغییر نام ستون‌ها به فارسی برای نمایش بهتر
columns_fa = {
    'Residential': 'بخش خانگی',
    'Commercial': 'بخش تجاری',
    'Industrial': 'بخش صنعتی',
    'Power Generation': 'تولید برق',
    'Total Consumption': 'مصرف کل',
    'Price': 'قیمت گاز (دلار/Mcf)'
}
df = df.rename(columns=columns_fa)

# استخراج ویژگی‌های زمانی
df['ماه'] = df.index.month
df['سال'] = df.index.year
df['فصل'] = df.index.quarter

# --- ایجاد مدل پیش‌بینی قیمت ---
features = ['ماه', 'بخش خانگی', 'بخش صنعتی', 'تولید برق']
target = 'قیمت گاز (دلار/Mcf)'

model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(df[features], df[target])

# --- تحلیل SHAP ---
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(df[features])

plt.figure(figsize=(12, 8))
shap.summary_plot(shap_values, df[features], show=False)
plt.title(format_persian('توزیع تأثیر ویژگی‌ها بر قیمت گاز (SHAP)'), fontsize=14)
plt.tight_layout()
plt.savefig('shap_plot.png', dpi=300)
plt.show()

# --- نمودار وابستگی جزئی (PDP) ---
plt.figure(figsize=(10, 6))
PartialDependenceDisplay.from_estimator(
    model,
    df[features],
    features=['ماه'],
    kind='average',
    line_kw={'linewidth': 3, 'color': 'red'}
)
plt.title(format_persian('تأثیر ماه سال بر قیمت گاز (PDP)'), fontsize=14)
plt.xlabel(format_persian('ماه سال'), fontsize=12)
plt.ylabel(format_persian('تغییرات قیمت گاز (دلار/Mcf)'), fontsize=12)
plt.grid(True, alpha=0.3)
plt.savefig('pdp_plot.png', dpi=300)
plt.show()

# --- مدل GAM برای بخش خانگی ---
gam = LinearGAM(s(0, n_splines=12)).fit(df['ماه'], df['بخش خانگی'])

plt.figure(figsize=(10, 6))
plt.plot(gam.partial_dependence(term=0, X=range(1, 13)), 'b-', linewidth=3)
plt.title(format_persian('الگوی ماهانه مصرف گاز در بخش خانگی (GAM)'), fontsize=14)
plt.xlabel(format_persian('ماه سال'), fontsize=12)
plt.ylabel(format_persian('مصرف گاز (بیلیون فوت مکعب)'), fontsize=12)
plt.grid(True, alpha=0.3)
plt.xticks(range(1, 13))
plt.savefig('gam_plot.png', dpi=300)
plt.show()

# --- تحلیل همبستگی و روندهای زمانی ---
plt.figure(figsize=(14, 10))

# نمودار روند زمانی
plt.subplot(2, 1, 1)
df['مصرف کل'].plot(linewidth=1.5, color='navy')
plt.title(format_persian('روند مصرف گاز طبیعی در آمریکا (2001-2025)'), fontsize=14)
plt.xlabel(format_persian('تاریخ'), fontsize=12)
plt.ylabel(format_persian('مصرف کل (بیلیون فوت مکعب)'), fontsize=12)
plt.grid(True, alpha=0.3)

# نمودار همبستگی
plt.subplot(2, 1, 2)
corr_matrix = df[['بخش خانگی', 'بخش تجاری', 'بخش صنعتی', 'تولید برق', 'قیمت گاز (دلار/Mcf)']].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title(format_persian('ماتریس همبستگی بین بخش‌های مصرف و قیمت'), fontsize=14)
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.savefig('correlation_plot.png', dpi=300)
plt.show()

# --- تحلیل فصلی مصرف ---
seasonal_analysis = df.groupby('فصل')[['بخش خانگی', 'بخش تجاری', 'بخش صنعتی']].mean()

plt.figure(figsize=(12, 6))
seasonal_analysis.plot(kind='bar', width=0.8)
plt.title(format_persian('میانگین مصرف فصلی گاز طبیعی (2001-2025)'), fontsize=14)
plt.xlabel(format_persian('فصل'), fontsize=12)
plt.ylabel(format_persian('میانگین مصرف (بیلیون فوت مکعب)'), fontsize=12)
plt.xticks([0, 1, 2, 3],
           [format_persian('زمستان'),
            format_persian('بهار'),
            format_persian('تابستان'),
            format_persian('پاییز')],
           rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.legend(title=format_persian('بخش مصرف'))
plt.tight_layout()
plt.savefig('seasonal_plot.png', dpi=300)
plt.show()

# --- تحلیل رابطه قیمت و مصرف ---
plt.figure(figsize=(12, 8))
plt.scatter(df['مصرف کل'], df['قیمت گاز (دلار/Mcf)'],
            c=df['ماه'], cmap='viridis', alpha=0.6, s=80)
plt.colorbar(label=format_persian('ماه سال'))
plt.title(format_persian('رابطه بین مصرف کل و قیمت گاز طبیعی'), fontsize=14)
plt.xlabel(format_persian('مصرف کل (بیلیون فوت مکعب)'), fontsize=12)
plt.ylabel(format_persian('قیمت گاز (دلار/Mcf)'), fontsize=12)
plt.grid(True, alpha=0.3)
plt.savefig('price_vs_consumption.png', dpi=300)
plt.show()
````

````python
# -*- coding: utf-8 -*-
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import IsolationForest
from sklearn.svm import OneClassSVM
from sklearn.metrics import roc_auc_score, f1_score, confusion_matrix
from tensorflow.keras.models import Model, Sequential
from tensorflow.keras.layers import Input, Dense, LSTM, Conv1D, MaxPooling1D, Flatten, Reshape, Dropout, TimeDistributed
from tensorflow.keras.callbacks import EarlyStopping
from arabic_reshaper import reshape
from bidi.algorithm import get_display
import warnings
warnings.filterwarnings('ignore')

# --- تنظیمات اولیه برای پشتیبانی از فارسی ---
plt.rcParams['font.family'] = 'B Nazanin'
plt.rcParams['font.size'] = 12

def prepare_arabic_text(text):
    reshaped_text = reshape(text)
    return get_display(reshaped_text)

# --- بارگذاری و پیش‌پردازش داده‌ها ---
def load_and_preprocess_data():
    # بارگذاری داده‌ها (فرض بر وجود فایل‌های CSV)
    train_normal = pd.read_csv('fault_free_training.csv')
    test_data = pd.read_csv('fault_free_testing.csv')

    # استخراج ویژگی‌ها و برچسب‌ها
    X_train = train_normal.iloc[:, 3:55].values
    X_test = test_data.iloc[:, 3:55].values
    y_test = test_data.iloc[:, 0].values

    # نرمال‌سازی داده‌ها
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # تبدیل برچسب‌ها به باینری (ناهنجاری/طبیعی)
    y_test_binary = np.where(y_test > 0, 1, 0)

    return X_train_scaled, X_test_scaled, y_test_binary

# --- مدل‌های تشخیص ناهنجاری ---
class AnomalyDetector:
    # 1. Isolation Forest
    def isolation_forest_model(self, X_train, X_test, y_test):
        model = IsolationForest(n_estimators=150, contamination=0.05, random_state=42)
        model.fit(X_train)
        scores = -model.decision_function(X_test)
        predictions = np.where(scores > np.percentile(scores, 95), 1, 0)
        return scores, predictions

    # 2. Autoencoder
    def autoencoder_model(self, X_train, X_test, y_test):
        input_dim = X_train.shape[1]

        # معماری مدل
        input_layer = Input(shape=(input_dim,))
        encoded = Dense(32, activation='relu')(input_layer)
        decoded = Dense(input_dim, activation='linear')(encoded)

        model = Model(input_layer, decoded)
        model.compile(optimizer='adam', loss='mse')

        # آموزش مدل
        history = model.fit(X_train, X_train,
                           epochs=100,
                           batch_size=32,
                           validation_split=0.1,
                           callbacks=[EarlyStopping(patience=5)],
                           verbose=0)

        # محاسبه خطای بازسازی
        reconstructions = model.predict(X_test)
        mse = np.mean(np.power(X_test - reconstructions, 2), axis=1)
        predictions = np.where(mse > np.percentile(mse, 95), 1, 0)
        return mse, predictions, history

    # 3. LSTM Autoencoder
    def lstm_autoencoder_model(self, X_train, X_test, y_test, timesteps=10):
        # تبدیل داده به فرمت سری زمانی
        X_train_ts = self.create_sequences(X_train, timesteps)
        X_test_ts = self.create_sequences(X_test, timesteps)

        # معماری مدل
        model = Sequential()
        model.add(LSTM(64, activation='relu', input_shape=(timesteps, X_train.shape[1]), return_sequences=True))
        model.add(LSTM(32, activation='relu', return_sequences=False))
        model.add(RepeatVector(timesteps))
        model.add(LSTM(32, activation='relu', return_sequences=True))
        model.add(LSTM(64, activation='relu', return_sequences=True))
        model.add(TimeDistributed(Dense(X_train.shape[1])))
        model.compile(optimizer='adam', loss='mse')

        # آموزش مدل
        history = model.fit(X_train_ts, X_train_ts,
                          epochs=50,
                          batch_size=32,
                          validation_split=0.1,
                          callbacks=[EarlyStopping(patience=3)],
                          verbose=0)

        # محاسبه خطای بازسازی
        reconstructions = model.predict(X_test_ts)
        mse = np.mean(np.power(X_test_ts - reconstructions, 2), axis=(1,2))
        predictions = np.where(mse > np.percentile(mse, 95), 1, 0)
        return mse, predictions, history

    # 4. One-Class SVM
    def one_class_svm_model(self, X_train, X_test, y_test):
        model = OneClassSVM(nu=0.05, kernel='rbf')
        model.fit(X_train)
        scores = -model.decision_function(X_test)
        predictions = np.where(scores > np.percentile(scores, 95), 1, 0)
        return scores, predictions

    # 5. CNN-LSTM Hybrid
    def cnn_lstm_model(self, X_train, X_test, y_test, timesteps=10):
        # تبدیل داده به فرمت سری زمانی
        X_train_ts = self.create_sequences(X_train, timesteps)
        X_test_ts = self.create_sequences(X_test, timesteps)

        # تبدیل برچسب‌ها
        y_test_ts = y_test[timesteps-1:]

        # معماری مدل
        model = Sequential()
        model.add(Conv1D(64, 3, activation='relu', input_shape=(timesteps, X_train.shape[1])))
        model.add(MaxPooling1D(2))
        model.add(LSTM(64, return_sequences=True))
        model.add(LSTM(32))
        model.add(Dense(1, activation='sigmoid'))
        model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

        # آموزش مدل
        history = model.fit(X_train_ts, np.zeros(len(X_train_ts)),  # آموزش بدون نظارت
                         epochs=50,
                         batch_size=32,
                         validation_split=0.1,
                         callbacks=[EarlyStopping(patience=3)],
                         verbose=0)

        # پیش‌بینی
        scores = model.predict(X_test_ts).flatten()
        predictions = np.where(scores > 0.5, 1, 0)
        return scores, predictions, history

    # 6. MTAD-GAN (ساده‌سازی شده)
    def mtad_gan_model(self, X_train, X_test, y_test, timesteps=10):
        # تبدیل داده به فرمت سری زمانی
        X_train_ts = self.create_sequences(X_train, timesteps)
        X_test_ts = self.create_sequences(X_test, timesteps)

        # معماری مدل
        generator = Sequential([
            LSTM(64, return_sequences=True, input_shape=(timesteps, X_train.shape[1])),
            LSTM(32),
            Dense(timesteps * X_train.shape[1]),
            Reshape((timesteps, X_train.shape[1]))
        ])

        discriminator = Sequential([
            LSTM(32, return_sequences=True, input_shape=(timesteps, X_train.shape[1])),
            LSTM(16),
            Dense(1, activation='sigmoid')
        ])

        discriminator.compile(optimizer='adam', loss='binary_crossentropy')
        discriminator.trainable = False

        gan_input = Input(shape=(timesteps, X_train.shape[1]))
        generated = generator(gan_input)
        validity = discriminator(generated)

        gan = Model(gan_input, [generated, validity])
        gan.compile(loss=['mse', 'binary_crossentropy'], optimizer='adam')

        # آموزش مدل (ساده‌سازی شده)
        valid = np.ones((len(X_train_ts), 1))
        fake = np.zeros((len(X_train_ts), 1))

        for epoch in range(30):
            # آموزش discriminator
            idx = np.random.randint(0, len(X_train_ts), len(X_train_ts))
            real_samples = X_train_ts[idx]
            generated_samples = generator.predict(real_samples)

            d_loss_real = discriminator.train_on_batch(real_samples, valid[idx])
            d_loss_fake = discriminator.train_on_batch(generated_samples, fake[idx])
            d_loss = 0.5 * np.add(d_loss_real, d_loss_fake)

            # آموزش generator
            g_loss = gan.train_on_batch(real_samples, [real_samples, valid[idx]])

        # محاسبه خطای بازسازی
        reconstructions = generator.predict(X_test_ts)
        mse = np.mean(np.power(X_test_ts - reconstructions, 2), axis=(1,2))
        predictions = np.where(mse > np.percentile(mse, 95), 1, 0)
        return mse, predictions

    # تابع کمکی برای ایجاد دنباله‌های زمانی
    def create_sequences(self, data, timesteps):
        sequences = []
        for i in range(len(data) - timesteps + 1):
            sequences.append(data[i:i+timesteps])
        return np.array(sequences)

# --- ارزیابی و مقایسه مدل‌ها ---
def evaluate_models(X_train, X_test, y_test):
    detector = AnomalyDetector()
    results = {}

    # 1. ارزیابی Isolation Forest
    scores, preds = detector.isolation_forest_model(X_train, X_test, y_test)
    results['Isolation Forest'] = {
        'AUC': roc_auc_score(y_test, scores),
        'F1': f1_score(y_test, preds)
    }

    # 2. ارزیابی Autoencoder
    scores, preds, history = detector.autoencoder_model(X_train, X_test, y_test)
    results['Autoencoder'] = {
        'AUC': roc_auc_score(y_test, scores),
        'F1': f1_score(y_test, preds),
        'History': history
    }

    # 3. ارزیابی LSTM Autoencoder
    scores, preds, history = detector.lstm_autoencoder_model(X_train, X_test, y_test)
    results['LSTM Autoencoder'] = {
        'AUC': roc_auc_score(y_test[:len(scores)], scores),
        'F1': f1_score(y_test[:len(scores)], preds),
        'History': history
    }

    # 4. ارزیابی One-Class SVM
    scores, preds = detector.one_class_svm_model(X_train, X_test, y_test)
    results['One-Class SVM'] = {
        'AUC': roc_auc_score(y_test, scores),
        'F1': f1_score(y_test, preds)
    }

    # 5. ارزیابی CNN-LSTM
    scores, preds, history = detector.cnn_lstm_model(X_train, X_test, y_test)
    results['CNN-LSTM'] = {
        'AUC': roc_auc_score(y_test[:len(scores)], scores),
        'F1': f1_score(y_test[:len(scores)], preds),
        'History': history
    }

    # 6. ارزیابی MTAD-GAN
    scores, preds = detector.mtad_gan_model(X_train, X_test, y_test)
    results['MTAD-GAN'] = {
        'AUC': roc_auc_score(y_test[:len(scores)], scores),
        'F1': f1_score(y_test[:len(scores)], preds)
    }

    return results

# --- ترسیم نتایج ---
def plot_results(results):
    # مقایسه عملکرد مدل‌ها
    models = list(results.keys())
    auc_scores = [results[model]['AUC'] for model in models]
    f1_scores = [results[model]['F1'] for model in models]

    fig, ax = plt.subplots(2, 1, figsize=(12, 10))

    # نمودار AUC
    ax[0].bar(models, auc_scores, color='skyblue')
    ax[0].set_title(prepare_arabic_text('مقایسه AUC مدل‌های مختلف'))
    ax[0].set_ylabel('AUC Score')
    ax[0].set_ylim(0.7, 1.0)

    # نمودار F1-score
    ax[1].bar(models, f1_scores, color='salmon')
    ax[1].set_title(prepare_arabic_text('مقایسه F1-Score مدل‌های مختلف'))
    ax[1].set_ylabel('F1 Score')
    ax[1].set_ylim(0.7, 1.0)

    plt.tight_layout()
    plt.savefig('model_comparison.png', dpi=300)
    plt.show()

    # نمودار یادگیری برای مدل‌های عمیق
    for model_name in ['Autoencoder', 'LSTM Autoencoder', 'CNN-LSTM']:
        if model_name in results and 'History' in results[model_name]:
            history = results[model_name]['History'].history
            plt.figure(figsize=(10, 5))
            plt.plot(history['loss'], label=prepare_arabic_text('خطای آموزش'))
            plt.plot(history['val_loss'], label=prepare_arabic_text('خطای اعتبارسنجی'))
            plt.title(prepare_arabic_text(f'منحنی یادگیری {model_name}'))
            plt.ylabel('Loss')
            plt.xlabel('Epoch')
            plt.legend()
            plt.savefig(f'{model_name}_learning_curve.png', dpi=300)
            plt.show()

# --- اجرای اصلی ---
if __name__ == "__main__":
    # بارگذاری و پیش‌پردازش داده‌ها
    X_train, X_test, y_test = load_and_preprocess_data()

    # ارزیابی مدل‌ها
    results = evaluate_models(X_train, X_test, y_test)

    # نمایش نتایج
    print("\n" + "-"*50)
    print(prepare_arabic_text("نتایج ارزیابی مدل‌ها:"))
    for model, metrics in results.items():
        print(f"{model}:")
        print(f"  AUC = {metrics['AUC']:.4f}")
        print(f"  F1 Score = {metrics['F1']:.4f}")
        print("-"*50)

    # ترسیم نمودارها
    plot_results(results)
````

````python
import numpy as np
import matplotlib.pyplot as plt
from deap import base, creator, tools, algorithms
import random

# =============================
# ۱. تعریف داده‌های واقعی پالایشگاه (مطابق مرجع ۱۳)
# =============================
# پارامترهای واحدهای تعمیراتی
units = [
    {"id": 1, "name": "برج تقطیر", "duration": 14, "cost": 85000, "window": (30, 90), "crew": 1},
    {"id": 2, "name": "ریفرمر کاتالیستی", "duration": 21, "cost": 120000, "window": (60, 150), "crew": 2},
    {"id": 3, "name": "هیدروکراکر", "duration": 18, "cost": 95000, "window": (40, 120), "crew": 1},
    {"id": 4, "name": "واحد تصفیه گاز", "duration": 10, "cost": 65000, "window": (20, 80), "crew": 1},
    {"id": 5, "name": "واحد آلکیل‌اسیون", "duration": 12, "cost": 75000, "window": (50, 110), "crew": 2}
]

# منابع محدود (تعداد تیم‌های تعمیراتی)
available_crews = {1: 2, 2: 1}  # نوع ۱: ۲ تیم، نوع ۲: ۱ تیم

# هزینه توقف تولید (هزار دلار در روز)
downtime_cost_per_day = 25000

# محدودیت زمانی کل (روز)
time_horizon = 180

# =============================
# ۲. تنظیمات الگوریتم ژنتیک
# =============================
# تعریف ساختار کروموزوم
creator.create("FitnessMin", base.Fitness, weights=(-1.0,))
creator.create("Individual", list, fitness=creator.FitnessMin)

toolbox = base.Toolbox()

# تولید کروموزوم (زمان شروع هر کار)
toolbox.register("attr_start", random.randint, 0, time_horizon)
toolbox.register("individual", tools.initRepeat, creator.Individual,
                 toolbox.attr_start, n=len(units))
toolbox.register("population", tools.initRepeat, list, toolbox.individual)

# تابع محاسبه هزینه
def calculate_cost(individual):
    total_cost = 0
    crew_utilization = {crew_type: [] for crew_type in available_crews}

    # محاسبه هزینه تعمیرات و بررسی محدودیت پنجره زمانی
    for i, start in enumerate(individual):
        unit = units[i]
        end = start + unit["duration"]

        # جریمه خارج شدن از پنجره زمانی
        if start < unit["window"][0] or end > unit["window"][1]:
            total_cost += 50000  # جریمه سنگین

        # هزینه تعمیرات
        total_cost += unit["cost"]

        # هزینه توقف تولید (متناسب با مدت تعمیر)
        total_cost += unit["duration"] * downtime_cost_per_day

        # ردیابی استفاده از منابع
        crew_utilization[unit["crew"]].append((start, end))

    # بررسی محدودیت منابع
    for crew_type, intervals in crew_utilization.items():
        intervals.sort(key=lambda x: x[0])

        # بررسی تداخل زمانی
        for j in range(1, len(intervals)):
            if intervals[j][0] < intervals[j-1][1]:
                total_cost += 75000  # جریمه تداخل منابع

        # بررسی تعداد تیم‌های در دسترس
        if len(intervals) > available_crews[crew_type]:
            total_cost += 100000  # جریمه تجاوز از منابع

    return total_cost,

# ثبت توابع الگوریتم
toolbox.register("evaluate", calculate_cost)
toolbox.register("mate", tools.cxTwoPoint)
toolbox.register("mutate", tools.mutUniformInt, low=0, up=time_horizon, indpb=0.2)
toolbox.register("select", tools.selTournament, tournsize=3)

# =============================
# ۳. اجرای الگوریتم ژنتیک
# =============================
def main():
    # پارامترهای الگوریتم
    population_size = 50
    num_generations = 200
    cx_prob = 0.7
    mut_prob = 0.3

    # ایجاد جمعیت اولیه
    pop = toolbox.population(n=population_size)

    # ردیابی بهترین جواب
    hof = tools.HallOfFame(1)
    stats = tools.Statistics(lambda ind: ind.fitness.values[0])
    stats.register("min", np.min)
    stats.register("avg", np.mean)

    # اجرای الگوریتم
    pop, log = algorithms.eaSimple(
        pop, toolbox, cxpb=cx_prob, mutpb=mut_prob,
        ngen=num_generations, stats=stats, halloffame=hof, verbose=True
    )

    return pop, log, hof

if __name__ == "__main__":
    best_pop, logbook, best_individual = main()
    best_schedule = best_individual[0]

    # =============================
    # ۴. نمایش نتایج
    # =============================
    print("\nبهترین زمان‌بندی تعمیرات:")
    print("=" * 50)
    for i, start in enumerate(best_schedule):
        unit = units[i]
        print(f"{unit['name']}: شروع روز {start} - پایان روز {start + unit['duration']}")

    print("\nهزینه کل بهینه‌سازی شده:", calculate_cost(best_schedule)[0], "هزار دلار")

    # نمودار همگرایی
    gen = logbook.select("gen")
    min_fit = logbook.select("min")
    avg_fit = logbook.select("avg")

    plt.figure(figsize=(10, 6))
    plt.plot(gen, min_fit, 'b-', label="بهترین هزینه")
    plt.plot(gen, avg_fit, 'r-', label="میانگین هزینه")
    plt.xlabel("نسل")
    plt.ylabel("هزینه کل (هزار دلار)")
    plt.title("همگرایی الگوریتم ژنتیک")
    plt.legend()
    plt.grid()
    plt.savefig('convergence.png', dpi=300)
    plt.show()

    # نمایش زمان‌بندی
    plt.figure(figsize=(12, 8))
    for i, start in enumerate(best_schedule):
        plt.barh(units[i]['name'], units[i]['duration'], left=start,
                 color=plt.cm.tab20(i), edgecolor='black')

    plt.xlabel("زمان (روز)")
    plt.ylabel("واحدهای تعمیراتی")
    plt.title("زمان‌بندی بهینه تعمیرات")
    plt.grid(axis='x')
    plt.tight_layout()
    plt.savefig('schedule.png', dpi=300)
    plt.show()
````
