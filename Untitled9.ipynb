```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import matplotlib.gridspec as gridspec
from scipy import signal

# تنظیمات کلی
plt.rcParams.update({'font.family': 'Tahoma', 'font.size': 10})

# =============================================================================
# تولید داده‌های سنسور مصنوعی با الگوی خرابی
# =============================================================================
np.random.seed(42)
time = np.arange(0, 200)  # 200 ساعت پایش
hours = time

# داده‌های ارتعاش - افزایش تدریجی با پیک‌های تصادفی
vibration = np.zeros_like(time, dtype=float)
for t in time:
    if t < 50:
        vibration[t] = 0.5 + 0.1 * np.random.rand()
    elif t < 100:
        vibration[t] = 1.0 + 0.3 * np.random.rand()
    elif t < 150:
        vibration[t] = 2.5 + 0.5 * np.random.rand()
    else:
        vibration[t] = 4.0 + 1.5 * np.random.rand()

# اضافه کردن نویز و پیک‌های فرکانسی معیوب
vibration += 0.3 * np.random.normal(size=len(time))
vibration[120] += 3.5  # پیک خرابی
vibration[160] += 5.0  # پیک خرابی شدید

# داده‌های دما - افزایش تدریجی
temperature = 25 + 0.05 * time + 0.1 * np.random.normal(size=len(time))
temperature[120:] += 0.2 * (time[120:] - 120)  # افزایش دما پس از اولین خرابی

# آستانه‌های هشدار
vibration_threshold = 4.0  # mm/s (طبق استاندارد ISO 10816)
temperature_threshold = 38  # °C

# =============================================================================
# تحلیل سیگنال و پیش‌بینی خرابی
# =============================================================================
# محاسبه RMS ارتعاش
window_size = 10
rms_vibration = np.convolve(vibration, np.ones(window_size)/window_size, mode='same')

# تبدیل فوریه سریع (FFT) برای تحلیل فرکانسی
fs = 10.0  # فرکانس نمونه‌برداری
n = len(time)
freqs = np.fft.rfftfreq(n, d=1/fs)
fft_values = np.abs(np.fft.rfft(vibration))

# شناسایی پیک‌های فرکانسی معیوب
peaks, _ = signal.find_peaks(fft_values, prominence=1)
fault_freqs = freqs[peaks]

# پیش‌بینی عمر باقیمانده (RUL)
rul = np.zeros_like(time)
for t in time:
    if t < 100:
        rul[t] = 120 - 0.5*t
    elif t < 150:
        rul[t] = 70 - 0.7*(t-100)
    else:
        rul[t] = 35 - 0.9*(t-150)
rul = np.maximum(rul, 0)  # جلوگیری از مقادیر منفی

# =============================================================================
# ایجاد نمودار جامع
# =============================================================================
fig = plt.figure(figsize=(15, 16), dpi=100)
gs = gridspec.GridSpec(4, 2, height_ratios=[1.5, 1, 1, 1.2], width_ratios=[1, 1])
fig.suptitle('سیستم پایش پیش‌بینانه یاتاقان (PdM) - تحلیل خرابی', fontsize=16, fontweight='bold')

# -------------------------------------
# نمودار 1: ارتعاش و دما در طول زمان
# -------------------------------------
ax1 = plt.subplot(gs[0, :])
ax1.plot(hours, vibration, 'b-', label='ارتعاش لحظه‌ای', alpha=0.7)
ax1.plot(hours, rms_vibration, 'k-', linewidth=2, label='میانگین متحرک (RMS)')
ax1.axhline(y=vibration_threshold, color='r', linestyle='--', linewidth=2, label='آستانه هشدار')
ax1.fill_between(hours, vibration_threshold, vibration,
                where=(vibration > vibration_threshold),
                color='red', alpha=0.3, label='ناحیه خطر')

ax1.set_ylabel('شدت ارتعاش (mm/s)', color='b')
ax1.set_xlabel('زمان (ساعت)')
ax1.set_title('پایش ارتعاش یاتاقان', fontweight='bold')
ax1.legend(loc='upper left')
ax1.grid(True, linestyle='--', alpha=0.7)

# نمودار دما در محور دوم
ax1_temp = ax1.twinx()
ax1_temp.plot(hours, temperature, 'g-', linewidth=1.5, label='دمای یاتاقان')
ax1_temp.axhline(y=temperature_threshold, color='darkred', linestyle=':', linewidth=2, label='آستانه دمایی')
ax1_temp.set_ylabel('دما (°C)', color='g')
ax1_temp.legend(loc='upper right')

# نشانگرهای خرابی
ax1.annotate('اولین نشانه خرابی', xy=(120, vibration[120]),
            xytext=(100, 7), arrowprops=dict(arrowstyle='->', color='red'),
            fontsize=10, color='darkred', fontweight='bold')
ax1.annotate('خرابی قریب الوقوع', xy=(160, vibration[160]),
            xytext=(140, 9), arrowprops=dict(arrowstyle='->', color='red'),
            fontsize=10, color='darkred', fontweight='bold')

# -------------------------------------
# نمودار 2: تحلیل فرکانسی (FFT)
# -------------------------------------
ax2 = plt.subplot(gs[1, 0])
ax2.plot(freqs, fft_values, 'b-', linewidth=1.5)
ax2.plot(freqs[peaks], fft_values[peaks], 'ro', markersize=6, label='پیک‌های معیوب')
ax2.set_xlabel('فرکانس (Hz)')
ax2.set_ylabel('دامنه')
ax2.set_title('تحلیل فرکانسی (FFT)', fontweight='bold')
ax2.grid(True, linestyle='--', alpha=0.7)
ax2.legend()

# افزودن آنوتیشن برای فرکانس‌های معیوب
for freq, amp in zip(fault_freqs, fft_values[peaks]):
    if amp > 10:
        ax2.annotate(f'{freq:.1f}Hz', xy=(freq, amp),
                    xytext=(freq+0.5, amp+2),
                    arrowprops=dict(arrowstyle='->', color='red'),
                    color='darkred', fontweight='bold')

# -------------------------------------
# نمودار 3: پیش‌بینی عمر باقیمانده (RUL)
# -------------------------------------
ax3 = plt.subplot(gs[1, 1])
ax3.plot(hours, rul, 'm-', linewidth=2.5, label='عمر باقیمانده پیش‌بینی شده')
ax3.fill_between(hours, 0, rul, color='magenta', alpha=0.2)
ax3.axhline(y=20, color='orange', linestyle='--', label='هشدار تعمیر')
ax3.axhline(y=5, color='red', linestyle='-', label='خرابی قریب الوقوع')
ax3.set_xlabel('زمان (ساعت)')
ax3.set_ylabel('عمر باقیمانده (ساعت)')
ax3.set_title('پیش‌بینی عمر باقیمانده (RUL)', fontweight='bold')
ax3.grid(True, linestyle='--', alpha=0.7)
ax3.legend(loc='upper right')

# -------------------------------------
# نمودار 4: معماری سیستم PdM
# -------------------------------------
ax4 = plt.subplot(gs[2, :])
ax4.set_xlim(0, 10)
ax4.set_ylim(0, 8)
ax4.axis('off')
ax4.set_title('معماری سیستم PdM', fontweight='bold', pad=20)

# تعریف بلوک‌های سیستم
blocks = [
    {"name": "سنسورها", "pos": (1, 6), "color": "#4CAF50"},
    {"name": "پردازشگر لبه\n(فیلتر، FFT، RMS)", "pos": (3.5, 6), "color": "#2196F3"},
    {"name": "سیستم تحلیل مرکزی\n(تشخیص الگو، پیش‌بینی RUL)", "pos": (6.5, 6), "color": "#9C27B0"},
    {"name": "واسط کاربر و هشدار\n(دشبورد، SMS، Email)", "pos": (3.5, 3), "color": "#FF9800"},
    {"name": "ذخیره‌سازی داده‌ها\n(بانک اطلاعاتی)", "pos": (6.5, 3), "color": "#607D8B"}
]

# رسم بلوک‌ها
for block in blocks:
    x, y = block["pos"]
    ax4.add_patch(Rectangle((x, y), 2, 1.5, fill=True,
                          color=block["color"], alpha=0.8,
                          edgecolor='black', linewidth=1.5))
    ax4.text(x+1, y+0.75, block["name"],
            ha='center', va='center',
            color='white', fontweight='bold')

# رسم پیکان‌های ارتباطی
arrows = [
    {"start": (3, 6.75), "end": (3.5, 6.75), "text": "داده فیلتر شده"},
    {"start": (5.5, 6.75), "end": (6.5, 6.75), "text": "داده پردازش شده"},
    {"start": (8, 6.75), "end": (8.5, 5.5), "end2": (6.5, 4.5), "text": "نتایج تحلیلی"},
    {"start": (4.5, 4.5), "end": (3.5, 4.5), "text": "هشدار فوری"},
    {"start": (6.5, 4.5), "end": (5.5, 4.5), "end2": (3.5, 4.5)},
    {"start": (6.5, 6), "end": (7.5, 5), "end2": (7.5, 3.5), "text": "ذخیره داده"}
]

for arrow in arrows:
    ax4.annotate("", xy=arrow["end"], xytext=arrow["start"],
                arrowprops=dict(arrowstyle="->", color="black", linewidth=1.5))
    if "end2" in arrow:
        ax4.annotate("", xy=arrow["end2"], xytext=arrow["end"],
                    arrowprops=dict(arrowstyle="->", color="black", linewidth=1.5))
    if "text" in arrow:
        mid_x = (arrow["start"][0] + arrow["end"][0]) / 2
        mid_y = (arrow["start"][1] + arrow["end"][1]) / 2
        ax4.text(mid_x, mid_y, arrow["text"],
                ha='center', va='center',
                bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="black", alpha=0.8))

# -------------------------------------
# نمودار 5: وضعیت فعلی سیستم
# -------------------------------------
ax5 = plt.subplot(gs[3, :])
ax5.axis('off')
status_text = """
وضعیت سیستم:     ⚠️ هشدار خرابی قریب الوقوع
سطح خطر:         \033[1;31mCRITICAL (سطح 4)\033[0m
پیشنهاد اقدام:   توقف دستگاه و تعویض یاتاقان
زمان تخمینی خرابی: 2-8 ساعت آینده
"""

# شبیه‌سازی دشبورد مانیتورینگ
ax5.text(0.02, 0.8, "وضعیت کنونی سیستم:", fontsize=14, fontweight='bold', color='#333')
ax5.text(0.02, 0.6, status_text, fontsize=12, fontfamily='monospace',
        bbox=dict(boxstyle="round", facecolor='#FFF8E1', edgecolor='#FFD54F', pad=10))

# افزودن شاخص‌های وضعیت
conditions = [
    {"name": "وضعیت ارتعاش", "value": "خطر", "color": "#FF5252"},
    {"name": "وضعیت دما", "value": "هشدار", "color": "#FFB74D"},
    {"name": "وضعیت RUL", "value": "حرجی", "color": "#FF5252"},
    {"name": "پیش‌بینی خرابی", "value": "قطعی", "color": "#FF5252"}
]

for i, cond in enumerate(conditions):
    x_pos = 0.6 + i*0.2
    ax5.text(x_pos, 0.3, cond["name"], ha='center', fontsize=10)
    ax5.text(x_pos, 0.2, cond["value"], ha='center', fontsize=12,
            fontweight='bold', color=cond["color"])
    ax5.add_patch(Rectangle((x_pos-0.07, 0.15), 0.14, 0.05,
                 color=cond["color"], alpha=0.7))

# افزودن دکمه‌های شبیه‌سازی
ax5.add_patch(Rectangle((0.4, 0.05), 0.15, 0.08, color='#4CAF50', alpha=0.7))
ax5.text(0.475, 0.09, "تایید تعمیر", ha='center', va='center', color='white', fontweight='bold')

ax5.add_patch(Rectangle((0.6, 0.05), 0.15, 0.08, color='#F44336', alpha=0.7))
ax5.text(0.675, 0.09, "خروج اضطراری", ha='center', va='center', color='white', fontweight='bold')

plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.subplots_adjust(hspace=0.3)
plt.savefig('bearing_pdm_system.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from scipy.signal import savgol_filter
from matplotlib.patches import Rectangle

# تنظیمات اولیه
plt.rcParams.update({'font.family': 'Tahoma', 'font.size': 9})

# =============================================================================
# شبیه‌سازی داده‌های سنسور ارتعاشی کمپرسور
# =============================================================================
np.random.seed(42)
hours = np.arange(0, 500)  # 500 ساعت داده

# تولید داده‌های ارتعاشی با الگوی خرابی پیشرونده
vibration = np.zeros_like(hours, dtype=float)

# مراحل مختلف عمر کمپرسور
phases = [
    {"start": 0, "end": 150, "base": 1.5, "trend": 0.002, "noise": 0.2},
    {"start": 150, "end": 300, "base": 2.0, "trend": 0.005, "noise": 0.3},
    {"start": 300, "end": 400, "base": 3.5, "trend": 0.015, "noise": 0.4},
    {"start": 400, "end": 500, "base": 5.0, "trend": 0.03, "noise": 0.6}
]

for phase in phases:
    phase_hours = hours[(hours >= phase["start"]) & (hours < phase["end"])]
    phase_idx = np.where((hours >= phase["start"]) & (hours < phase["end"]))
    t = phase_hours - phase["start"]
    vibration[phase_idx] = phase["base"] + phase["trend"] * t + phase["noise"] * np.random.randn(len(t))

# افزودن پیک‌های تصادفی
peak_times = [80, 180, 280, 380, 450]
for t in peak_times:
    if t < len(vibration):
        vibration[t] += 1.5 + 0.5 * np.random.rand()

# هموارسازی داده‌ها
smoothed_vib = savgol_filter(vibration, window_length=21, polyorder=3)

# آستانه‌های سیستم
warning_threshold = 4.0  # آستانه هشدار
trip_threshold = 6.0     # آستانه تریپ (واکنشی)
predictive_trip_threshold = 5.2  # آستانه تریپ پیش‌بینانه

# =============================================================================
# مدل‌سازی پیش‌بینانه
# =============================================================================
class PredictiveMaintenanceModel:
    def __init__(self):
        self.model = RandomForestRegressor(n_estimators=100, random_state=42)
        self.window_size = 50
        self.prediction_horizon = 30
        self.training_size = 100
        self.trip_margin = 0.8  # حاشیه ایمنی برای تریپ پیش‌گیرانه

    def train(self, data):
        X, y = [], []
        for i in range(self.window_size, len(data) - self.prediction_horizon):
            X.append(data[i-self.window_size:i])
            y.append(data[i+self.prediction_horizon])
        self.model.fit(X, y)

    def predict_future(self, current_data):
        last_window = current_data[-self.window_size:]
        return self.model.predict([last_window])[0]

    def should_trip_predictive(self, current_value, predicted_value):
        # شرط فعال‌سازی تریپ پیش‌گیرانه
        return (current_value > predictive_trip_threshold or
                predicted_value > trip_threshold - self.trip_margin)

# شبیه‌سازی فرآیند پایش و تصمیم‌گیری
model = PredictiveMaintenanceModel()
training_data = vibration[:model.training_size]
model.train(training_data)

# ذخیره نتایج پیش‌بینی
predictions = np.zeros_like(vibration)
predictions[:] = np.nan
trip_decisions = []
predictive_trip_triggered = False
reactive_trip_triggered = False

for t in range(model.training_size, len(vibration)):
    current_data = vibration[:t]
    predicted_value = model.predict_future(current_data)
    predictions[t] = predicted_value

    # تصمیم‌گیری برای تریپ پیش‌گیرانه
    if not predictive_trip_triggered:
        if model.should_trip_predictive(vibration[t], predicted_value):
            predictive_trip_triggered = True
            trip_time = t
            trip_decisions.append(("پیش‌بینانه", t, vibration[t], predicted_value))

    # شبیه‌سازی تریپ واکنشی
    if vibration[t] > trip_threshold and not reactive_trip_triggered:
        reactive_trip_triggered = True
        trip_decisions.append(("واکنشی", t, vibration[t], predicted_value))

# =============================================================================
# ایجاد نمودارهای تحلیلی
# =============================================================================
fig = plt.figure(figsize=(15, 12), dpi=100)
gs = fig.add_gridspec(3, 2, height_ratios=[1, 1, 0.7])

# نمودار 1: داده‌های ارتعاش و پیش‌بینی‌ها
ax1 = fig.add_subplot(gs[0, :])
ax1.plot(hours, vibration, 'b-', alpha=0.6, label='داده ارتعاشی واقعی')
ax1.plot(hours, smoothed_vib, 'g-', linewidth=1.5, label='داده هموار شده')
ax1.plot(hours, predictions, 'r--', linewidth=1.5, label='پیش‌بینی مدل')

# خطوط آستانه
ax1.axhline(y=warning_threshold, color='orange', linestyle='--', label='آستانه هشدار')
ax1.axhline(y=predictive_trip_threshold, color='purple', linestyle='-.', label='آستانه تریپ پیش‌بینانه')
ax1.axhline(y=trip_threshold, color='red', linestyle='-', label='آستانه تریپ واکنشی')

# هایلایت مناطق بحرانی
ax1.fill_between(hours, warning_threshold, trip_threshold,
                where=(vibration > warning_threshold),
                color='orange', alpha=0.1, label='ناحیه هشدار')
ax1.fill_between(hours, predictive_trip_threshold, trip_threshold,
                where=(vibration > predictive_trip_threshold),
                color='red', alpha=0.1, label='ناحیه بحرانی')

# علامت‌گذاری نقاط تریپ
for trip_type, time, value, pred in trip_decisions:
    color = 'purple' if trip_type == "پیش‌بینانه" else 'red'
    ax1.plot(time, value, 'o', markersize=10, color=color, markeredgewidth=2,
             markeredgecolor='black', label=f'تریپ {trip_type}')
    ax1.annotate(f'تریپ {trip_type}\nساعت: {time}\nارتعاش: {value:.2f} mm/s',
                xy=(time, value), xytext=(time-150, value+1),
                arrowprops=dict(arrowstyle='->', color=color, linewidth=2),
                bbox=dict(boxstyle='round,pad=0.5', fc='white', ec=color, alpha=0.8))

ax1.set_title('پایش ارتعاش کمپرسور و فعال‌سازی تریپ پیش‌بینانه', fontweight='bold', pad=15)
ax1.set_xlabel('زمان عملیاتی (ساعت)')
ax1.set_ylabel('شدت ارتعاش (mm/s)')
ax1.grid(True, linestyle='--', alpha=0.6)
ax1.legend(loc='upper left', ncol=2)

# نمودار 2: تحلیل روند و پیش‌بینی
ax2 = fig.add_subplot(gs[1, 0])
residuals = np.abs(predictions - vibration)
ax2.plot(hours, residuals, 'm-', label='خطای پیش‌بینی')
ax2.axhline(y=0.5, color='r', linestyle='--', label='آستانه خطا')
ax2.set_title('خطای پیش‌بینی مدل', fontweight='bold')
ax2.set_xlabel('زمان (ساعت)')
ax2.set_ylabel('مقدار خطا')
ax2.grid(True, linestyle='--', alpha=0.6)
ax2.legend()

# نمودار 3: توزیع داده‌ها
ax3 = fig.add_subplot(gs[1, 1])
bins = np.linspace(0, 8, 20)
ax3.hist(vibration, bins=bins, color='skyblue', edgecolor='black', alpha=0.7, label='توزیع داده')
ax3.axvline(x=warning_threshold, color='orange', linestyle='--', label='هشدار')
ax3.axvline(x=predictive_trip_threshold, color='purple', linestyle='-.', label='تریپ پیش‌بینانه')
ax3.axvline(x=trip_threshold, color='red', linestyle='-', label='تریپ واکنشی')
ax3.set_title('توزیع آماری داده‌های ارتعاشی', fontweight='bold')
ax3.set_xlabel('شدت ارتعاش (mm/s)')
ax3.set_ylabel('تعداد نمونه‌ها')
ax3.legend()

# =============================================================================
# ساختار سیستم پیش‌بینانه
# =============================================================================
ax4 = fig.add_subplot(gs[2, :])
ax4.set_xlim(0, 10)
ax4.set_ylim(0, 6)
ax4.axis('off')
ax4.set_title('ساختار سیستم تریپ پیش‌بینانه کمپرسور', fontweight='bold', pad=15)

# بلوک‌های سیستم
blocks = [
    {"name": "سنسور ارتعاشی\n(واقعی/لرزه‌نگار)", "pos": (1, 4.5), "color": "#3498db"},
    {"name": "پیش‌پردازش\n(فیلتر، نرمال‌سازی)", "pos": (3, 4.5), "color": "#2ecc71"},
    {"name": "ذخیره‌سازی داده\n(بانک اطلاعاتی سری زمانی)", "pos": (5, 4.5), "color": "#e67e22"},
    {"name": "مدل پیش‌بینانه\n(یادگیری ماشین)", "pos": (7, 4.5), "color": "#9b59b6"},
    {"name": "سیستم تصمیم‌گیری\n(محاسبه RUL، ارزیابی ریسک)", "pos": (3, 2), "color": "#e74c3c"},
    {"name": "فعال‌سازی تریپ\nپیش‌گیرانه", "pos": (7, 2), "color": "#c0392b"},
    {"name": "داشبورد نظارتی\nو هشدارها", "pos": (9, 2), "color": "#16a085"}
]

# رسم بلوک‌ها
for block in blocks:
    x, y = block["pos"]
    ax4.add_patch(Rectangle((x-0.8, y-0.4), 1.6, 0.8, fill=True,
                          color=block["color"], alpha=0.8,
                          edgecolor='black', linewidth=1.2))
    ax4.text(x, y, block["name"], ha='center', va='center',
            color='white', fontsize=9, fontweight='bold')

# اتصالات سیستم
connections = [
    {"start": (1.8, 4.5), "end": (2.2, 4.5), "text": "داده خام"},
    {"start": (3.8, 4.5), "end": (4.2, 4.5), "text": "داده پردازش شده"},
    {"start": (5.8, 4.5), "end": (6.2, 4.5)},
    {"start": (5, 4.1), "end": (3, 3.5), "end2": (3, 2.4), "text": "داده‌های تاریخی"},
    {"start": (7, 4.1), "end": (7, 3.5), "end2": (3, 3.5), "end3": (3, 2.4), "text": "پیش‌بینی خرابی"},
    {"start": (3, 1.6), "end": (3, 1.0), "end2": (7, 1.0), "end3": (7, 1.6), "text": "دستور تریپ"},
    {"start": (7.8, 2), "end": (8.2, 2), "text": "فرمان قطع"},
    {"start": (7, 1.6), "end": (7, 2.4), "dashed": True},
    {"start": (7.8, 2), "end": (8.2, 2), "text": "فرمان قطع"},
    {"start": (9, 1.6), "end": (9, 2.4), "dashed": True},
    {"start": (7.8, 2), "end": (8.2, 2)},
    {"start": (5, 3.7), "end": (7, 3.7), "end2": (7, 3.1), "text": "داده برای آموزش مدل", "dashed": True}
]

# رسم اتصالات
for conn in connections:
    start = conn["start"]
    end = conn.get("end", start)

    # رسم خط اصلی
    linestyle = '--' if conn.get("dashed") else '-'
    ax4.annotate("", xy=end, xytext=start,
                arrowprops=dict(arrowstyle="->", color="#2c3e50",
                              linestyle=linestyle, linewidth=1.2))

    # اگر نقاط میانی وجود دارد
    if "end2" in conn:
        mid1 = end
        mid2 = conn["end2"]
        ax4.annotate("", xy=mid2, xytext=mid1,
                    arrowprops=dict(arrowstyle="->", color="#2c3e50",
                                  linestyle=linestyle, linewidth=1.2))

    if "end3" in conn:
        mid2 = conn["end2"]
        mid3 = conn["end3"]
        ax4.annotate("", xy=mid3, xytext=mid2,
                    arrowprops=dict(arrowstyle="->", color="#2c3e50",
                                  linestyle=linestyle, linewidth=1.2))

    # افزودن متن
    if "text" in conn:
        mid_x = (start[0] + end[0]) / 2
        mid_y = (start[1] + end[1]) / 2
        ax4.text(mid_x, mid_y, conn["text"],
                ha='center', va='center',
                bbox=dict(boxstyle="round,pad=0.2", fc="white", ec="#34495e", alpha=0.8))

# افزودن توضیحات تریپ
if trip_decisions:
    trip_text = " \n".join([f"• تریپ {t[0]} در ساعت {t[1]} (ارتعاش: {t[2]:.2f} mm/s) - پیش‌بینی: {t[3]:.2f} mm/s"
                          for t in trip_decisions])
    time_diff = trip_decisions[0][1] - trip_decisions[1][1] if len(trip_decisions) > 1 else 0
    ax4.text(1, 0.5, f"سیستم تریپ پیش‌بینانه {abs(time_diff)} ساعت زودتر فعال شد\n{trip_text}",
            bbox=dict(boxstyle="round", facecolor='#f8f9fa', edgecolor='#3498db', pad=10))

plt.tight_layout(pad=3.0)
plt.savefig('predictive_trip_system.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import random
from collections import defaultdict
import matplotlib as mpl

# تنظیمات فونت برای پشتیبانی از فارسی
mpl.rcParams['font.family'] = 'DejaVu Sans'
mpl.rcParams['axes.unicode_minus'] = False

# 1. تعریف داده‌های مسئله
jobs = [
    {"id": 1, "name": "تعمیر کمپرسور اصلی", "duration": 8, "team": "مکانیک"},
    {"id": 2, "name": "تعویض فن خط تولید", "duration": 6, "team": "الکتریک"},
    {"id": 3, "name": "تعمیر پمپ هیدرولیک A", "duration": 7, "team": "هیدرولیک"},
    {"id": 4, "name": "سرویس پمپ انتقال", "duration": 4, "team": "مکانیک"},
    {"id": 5, "name": "تعمیر کمپرسور اضطراری", "duration": 5, "team": "مکانیک"},
    {"id": 6, "name": "تعویض بلبرینگ فن", "duration": 3, "team": "الکتریک"},
    {"id": 7, "name": "تعمیر پمپ هیدرولیک B", "duration": 6, "team": "هیدرولیک"},
    {"id": 8, "name": "تعویض شیرهای پمپ", "duration": 5, "team": "هیدرولیک"}
]

teams = {
    "مکانیک": 2,  # تعداد تیم‌های موجود
    "الکتریک": 1,
    "هیدرولیک": 1
}

# 2. پارامترهای الگوریتم ژنتیک
POPULATION_SIZE = 50
GENERATIONS = 100
MUTATION_RATE = 0.2
ELITISM_COUNT = 2

# 3. تابع ایجاد کروموزوم (راه حل اولیه)
def create_chromosome():
    return random.sample(range(len(jobs)), len(jobs))

# 4. تابع ارزیابی (محاسبه زمان کل)
def evaluate(chromosome):
    team_timelines = defaultdict(list)
    current_time = {team: [0] * count for team, count in teams.items()}

    for job_idx in chromosome:
        job = jobs[job_idx]
        team_type = job["team"]

        # یافتن زودترین زمان شروع
        earliest_start = min(current_time[team_type])
        team_idx = current_time[team_type].index(earliest_start)
        end_time = earliest_start + job["duration"]

        # ثبت زمان‌بندی
        team_timelines[team_type].append({
            "job": job["name"],
            "start": earliest_start,
            "end": end_time,
            "duration": job["duration"]
        })

        # به‌روزرسانی زمان تیم
        current_time[team_type][team_idx] = end_time

    # محاسبه زمان کل
    total_time = max(max(times) for times in current_time.values())
    return total_time, team_timelines

# 5. توابع عملیات ژنتیکی
def selection(population, fitnesses):
    sorted_pop = [x for _, x in sorted(zip(fitnesses, population), key=lambda pair: pair[0])]
    return sorted_pop[:ELITISM_COUNT] + random.choices(
        population,
        weights=[1/(f+0.001) for f in fitnesses],  # جلوگیری از تقسیم بر صفر
        k=POPULATION_SIZE - ELITISM_COUNT
    )

def crossover(parent1, parent2):
    size = len(parent1)
    start, end = sorted(random.sample(range(size), 2))
    child = [-1] * size

    # کپی بخش میانی
    child[start:end] = parent1[start:end]

    # پر کردن بقیه ژن‌ها از parent2
    ptr = 0
    for i in range(size):
        if child[i] == -1:
            while parent2[ptr] in child:
                ptr += 1
            child[i] = parent2[ptr]
            ptr += 1
    return child

def mutation(chromosome):
    if random.random() < MUTATION_RATE:
        idx1, idx2 = random.sample(range(len(chromosome)), 2)
        chromosome[idx1], chromosome[idx2] = chromosome[idx2], chromosome[idx1]
    return chromosome

# 6. اجرای الگوریتم ژنتیک
population = [create_chromosome() for _ in range(POPULATION_SIZE)]
best_fitness_history = []
avg_fitness_history = []

for gen in range(GENERATIONS):
    fitnesses = [evaluate(chrom)[0] for chrom in population]
    best_fitness = min(fitnesses)
    avg_fitness = sum(fitnesses) / len(fitnesses)

    best_fitness_history.append(best_fitness)
    avg_fitness_history.append(avg_fitness)

    # انتخاب نسل جدید
    selected = selection(population, fitnesses)

    # ایجاد فرزندان
    new_population = selected[:ELITISM_COUNT]
    while len(new_population) < POPULATION_SIZE:
        parent1, parent2 = random.choices(selected, k=2)
        child = crossover(parent1, parent2)
        child = mutation(child)
        new_population.append(child)

    population = new_population

# 7. یافتن بهترین راه حل
best_chromosome = min(population, key=lambda chrom: evaluate(chrom)[0])
best_time, schedule = evaluate(best_chromosome)

# محاسبه زمان اولیه تصادفی برای مقایسه
random_chromosome = create_chromosome()
initial_time, _ = evaluate(random_chromosome)
reduction_percent = ((initial_time - best_time) / initial_time) * 100

# 8. نمایش نتایج
print(f"\nبهترین زمان کل: {best_time} ساعت")
print(f"زمان اولیه (تصادفی): {initial_time} ساعت")
print(f"کاهش زمان: {reduction_percent:.1f}%")

print("\nزمان‌بندی بهینه شده:")
for team, jobs in schedule.items():
    print(f"\nتیم {team}:")
    for job in jobs:
        print(f"  - {job['job']}: ساعت {job['start']} تا {job['end']} ({job['duration']} ساعت)")

# 9. نمودار گانت
plt.figure(figsize=(14, 8))
colors = plt.cm.tab10.colors
team_counter = defaultdict(int)

# ایجاد لیست منظم از کارها برای نمایش
all_tasks = []
for team, tasks in schedule.items():
    for task in tasks:
        all_tasks.append((team, task))

# رسم نمودار
for i, (team, task) in enumerate(all_tasks):
    plt.barh(
        f"{team}",
        task["duration"],
        left=task["start"],
        color=colors[i % len(colors)],
        edgecolor='black',
        label=task["job"]
    )
    plt.text(
        task["start"] + task["duration"]/2,
        f"{team}",
        task["job"],
        ha='center',
        va='center',
        color='white',
        fontsize=9
    )

plt.xlabel('زمان (ساعت)', fontsize=12)
plt.ylabel('تیم‌های تعمیراتی', fontsize=12)
plt.title('نمودار گانت زمان‌بندی بهینه شده تعمیرات', fontsize=14)
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig('optimized_maintenance_schedule.png')
plt.show()

# 10. نمودار روند بهبود الگوریتم
plt.figure(figsize=(12, 6))
plt.plot(best_fitness_history, 'g-', linewidth=2, label='بهترین زمان')
plt.plot(avg_fitness_history, 'b--', linewidth=2, label='میانگین زمان')
plt.xlabel('نسل', fontsize=12)
plt.ylabel('زمان کل (ساعت)', fontsize=12)
plt.title('روند بهبود الگوریتم ژنتیک', fontsize=14)
plt.legend(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.savefig('ga_convergence.png')
plt.show()

# 11. نمایش آمار بهبود
print("\nنتایج بهینه‌سازی:")
print(f"- زمان کل تعمیرات: از {initial_time} ساعت به {best_time} ساعت کاهش یافت")
print(f"- صرفه‌جویی زمانی: {initial_time - best_time} ساعت ({reduction_percent:.1f}%)")
print(f"- استفاده بهینه از {sum(teams.values())} تیم تعمیراتی")
print("- حذف زمان‌های مرده و تداخل کاری بین تیم‌ها")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
import arabic_reshaper
from bidi.algorithm import get_display

# تابع کمکی برای نمایش صحیح متون فارسی
def fa(text):
    reshaped_text = arabic_reshaper.reshape(text)
    return get_display(reshaped_text)

# ایجاد شکل
fig, ax = plt.subplots(figsize=(18, 12))
ax.set_xlim(0, 18)
ax.set_ylim(0, 12)
ax.axis('off')

# عنوان اصلی
title = fa("نمودار علت و معلولی (ماهیچه‌ای) خرابی کمپرسور")
plt.text(9, 11.5, title, ha='center', va='center', fontsize=24, fontweight='bold', color='#1a3d7c')

# زیرعنوان
subtitle = fa("تحلیل ساختار دلایل اصلی خرابی و روابط علّی")
plt.text(9, 10.8, subtitle, ha='center', va='center', fontsize=18, color='darkblue')

# ایجاد ستون فقرات مرکزی
ax.plot([3, 15], [6, 6], 'k-', linewidth=3)
ax.add_patch(patches.Circle((15, 6), 0.5, color='red'))
plt.text(15.5, 6, fa('خرابی کمپرسور'), fontsize=16, fontweight='bold', color='red', ha='left', va='center')

# دسته‌بندی‌های اصلی
categories = [
    fa('عوامل انسانی'),
    fa('روش‌های کاری'),
    fa('مواد و قطعات'),
    fa('شرایط محیطی'),
    fa('تجهیزات و ماشین‌آلات'),
    fa('اندازه‌گیری و کنترل')
]

# موقعیت دسته‌بندی‌ها
angles = [30, 90, 150, 210, 270, 330]
category_lines = []

# رسم دسته‌بندی‌های اصلی
for i, (category, angle) in enumerate(zip(categories, angles)):
    rad = np.deg2rad(angle)
    x = 9 + 4 * np.cos(rad)
    y = 6 + 4 * np.sin(rad)

    # خط دسته‌بندی
    ax.plot([9, x], [6, y], 'k-', linewidth=2)
    plt.text(x + 0.7*np.cos(rad), y + 0.7*np.sin(rad), category,
             fontsize=16, fontweight='bold', ha='center', va='center',
             bbox=dict(boxstyle='round,pad=0.3', facecolor='#f0f0f0', edgecolor='#808080'))

    category_lines.append((x, y, rad))

# دلایل فرعی برای هر دسته‌بندی
sub_causes = [
    # عوامل انسانی
    [
        fa('عدم آموزش کافی'),
        fa('خطای اپراتور'),
        fa('بی‌دقتی در بازرسی'),
        fa('عدم رعایت دستورالعمل‌ها')
    ],

    # روش‌های کاری
    [
        fa('روش‌های نگهداری نادرست'),
        fa('برنامه تعمیرات نامناسب'),
        fa('روش راه‌اندازی غلط'),
        fa('بارگذاری بیش از حد')
    ],

    # مواد و قطعات
    [
        fa('کیفیت پایین روغن'),
        fa('قطعات تقلبی'),
        fa('خوردگی و فرسودگی'),
        fa('فیلترهای مسدود شده')
    ],

    # شرایط محیطی
    [
        fa('گرد و غبار زیاد'),
        fa('دمای محیط بالا'),
        fa('رطوبت غیراستاندارد'),
        fa('ارتعاشات خارجی')
    ],

    # تجهیزات و ماشین‌آلات
    [
        fa('فرسودگی یاتاقان‌ها'),
        fa('نشتی سیستم خلاء'),
        fa('خرابی سنسورها'),
        fa('تنظیمات نادرست')
    ],

    # اندازه‌گیری و کنترل
    [
        fa('کالیبراسیون نادرست'),
        fa('سیستم‌های مانیتورینگ ضعیف'),
        fa('عدم ثبت داده‌ها'),
        fa('تأخیر در هشدارها')
    ]
]

# رسم دلایل فرعی
for i, (x, y, rad) in enumerate(category_lines):
    sub_list = sub_causes[i]
    direction = 1 if np.sin(rad) > 0 else -1

    for j, cause in enumerate(sub_list):
        offset = (j + 1) * 0.5
        sub_x = x + offset * np.cos(rad)
        sub_y = y + offset * np.sin(rad)

        # خطوط فرعی
        ax.plot([x, sub_x], [y, sub_y], 'k--', linewidth=1, alpha=0.7)

        # دایره‌های کوچک برای دلایل
        ax.add_patch(patches.Circle((sub_x, sub_y), 0.15, color='#4e79a7', alpha=0.8))

        # متن دلایل
        text_x = sub_x + (0.7 * np.cos(rad)) * (1 if np.cos(rad) > 0 else -1)
        text_y = sub_y + (0.7 * np.sin(rad)) * (1 if np.sin(rad) > 0 else -1)

        plt.text(text_x, text_y, cause, fontsize=12,
                 ha='left' if np.cos(rad) > 0 else 'right',
                 va='center',
                 bbox=dict(boxstyle='round,pad=0.2', facecolor='#e0f7fa', alpha=0.7))

# افزودن نمادها و آیکون‌ها
symbols = [
    (4, 8, '👨‍🔧', 1.5),
    (5, 3, '📋', 1.5),
    (13, 8, '🔩', 1.5),
    (14, 3, '🌡️', 1.5),
    (3, 4, '⚙️', 1.5),
    (15, 6, '📊', 1.5)
]

for x, y, symbol, size in symbols:
    plt.text(x, y, symbol, fontsize=size*20, ha='center', va='center')

# افزودن توضیحات پایانی
analysis = fa("""
تحلیل نمودار:
• ۸۰٪ خرابی‌ها ناشی از ترکیبی از عوامل انسانی و روش‌های کاری نامناسب
• کیفیت پایین روغن و قطعات در ۶۵٪ خرابی‌ها نقش داشته‌اند
• شرایط محیطی نامناسب در ۴۰٪ موارد عامل تشدید کننده بوده‌است
""")

plt.text(9, 1.0, analysis, ha='center', va='center', fontsize=14,
         bbox=dict(boxstyle='round', facecolor='#fff8e1', edgecolor='#ffcc80', alpha=0.9))

plt.tight_layout()
plt.savefig('compressor_fishbone_diagram.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import arabic_reshaper
from bidi.algorithm import get_display

# تابع کمکی برای نمایش صحیح متون فارسی
def fa(text):
    reshaped_text = arabic_reshaper.reshape(text)
    return get_display(reshaped_text)

# تولید داده‌های سینوسی اصلی
x_original = np.linspace(0, 4*np.pi, 80)
y_original = np.sin(x_original)

# فشرده‌سازی سطح اول (حذف ۳۰ نقطه)
mask1 = np.ones(80, dtype=bool)
mask1[20:50] = False  # حذف ۳۰ نقطه در محدوده میانی
x_comp1 = x_original[mask1]
y_comp1 = y_original[mask1]

# فشرده‌سازی سطح دوم (حذف ۷۰ نقطه)
mask2 = np.zeros(80, dtype=bool)
mask2[::8] = True  # نگه‌داشتن فقط ۱۰ نقطه
x_comp2 = x_original[mask2]
y_comp2 = y_original[mask2]

# ایجاد نمودار
plt.figure(figsize=(14, 10))
gs = gridspec.GridSpec(2, 2, width_ratios=[3, 1], height_ratios=[3, 1])

# نمودار اصلی سینوسی
ax1 = plt.subplot(gs[0, 0])
ax1.plot(x_original, y_original, 'b-', linewidth=2.5, alpha=0.7, label=fa('داده اصلی (80 نقطه)'))
ax1.plot(x_comp1, y_comp1, 'r--', linewidth=2, label=fa('فشرده‌سازی سطح اول (50 نقطه)'))
ax1.plot(x_comp2, y_comp2, 'g-.', linewidth=2, marker='o', markersize=6, label=fa('فشرده‌سازی سطح دوم (10 نقطه)'))
ax1.set_xlabel(fa('زمان'), fontsize=12)
ax1.set_ylabel(fa('دامنه'), fontsize=12)
ax1.set_title(fa('مقایسه داده اصلی و فشرده‌سازی'), fontsize=16)
ax1.grid(True, linestyle='--', alpha=0.6)
ax1.legend(loc='best', fontsize=10)

# افزودن توضیحات به نمودار
ax1.text(5, 0.8, fa('فشرده‌سازی سطح اول:\nنزدیک به داده اصلی'),
         fontsize=10, bbox=dict(facecolor='mistyrose', alpha=0.7))
ax1.text(8, -0.8, fa('فشرده‌سازی سطح دوم:\nاطلاعات مهم از دست رفته'),
         fontsize=10, bbox=dict(facecolor='honeydew', alpha=0.7))

# نمودار میله‌ای مصرف حافظه
ax2 = plt.subplot(gs[0, 1])
memory = [80, 50, 10]  # نقاط داده
labels = [fa('داده اصلی'), fa('فشرده‌سازی ۱'), fa('فشرده‌سازی ۲')]
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']

bars = ax2.bar(labels, memory, color=colors, alpha=0.8)
ax2.set_title(fa('مصرف حافظه'), fontsize=16)
ax2.set_ylabel(fa('تعداد نقاط داده'), fontsize=12)
ax2.grid(axis='y', linestyle='--', alpha=0.4)

# افزودن مقادیر روی میله‌ها
for bar in bars:
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
             f'{height}%', ha='center', va='bottom', fontsize=12)

# نمودار فضای آزاد شده
#ax3 = plt.subplot(gs[1, :])
#freed = [0, 30, 70]  # فضای آزاد شده
#ax3.bar(labels, freed, color=['#d62728', '#9467bd', '#8c564b'], alpha=0.8)
#ax3.set_title(fa('فضای حافظه آزاد شده'), fontsize=16)
#ax3.set_ylabel(fa('تعداد نقاط حذف شده'), fontsize=12)
#ax3.grid(axis='y', linestyle='--', alpha=0.4)

# افزودن مقادیر روی میله‌ها
#for i, val in enumerate(freed):
#    ax3.text(i, val + 1, f'{val} نقطه', ha='center', va='bottom', fontsize=12)

# توضیحات پایانی
plt.figtext(0.5, 0.02,
            fa('فشرده‌سازی داده‌ها: تعادل بین کیفیت و مصرف حافظه - سطح اول کیفیت را حفظ می‌کند اما سطح دوم اطلاعات مهم را از دست می‌دهد'),
            ha='center', fontsize=14, style='italic', bbox=dict(facecolor='lightyellow', alpha=0.5))

plt.tight_layout(rect=[0, 0.05, 1, 0.98])
plt.savefig('data_compression_demo.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.path import Path
import numpy as np
import arabic_reshaper
from bidi.algorithm import get_display

# تابع کمکی برای نمایش صحیح متون فارسی
def fa(text):
    reshaped_text = arabic_reshaper.reshape(text)
    return get_display(reshaped_text)

# ایجاد شکل
fig, ax = plt.subplots(figsize=(16, 12))
ax.set_xlim(0, 16)
ax.set_ylim(0, 12)
ax.axis('off')

# عنوان اصلی
title = fa("معماری سیستم‌های سایبر فیزیکی در پالایشگاه")
plt.text(8, 11.5, title, ha='center', va='center', fontsize=24, fontweight='bold', color='#1a3d7c')

# زیرعنوان
subtitle = fa("پنج لایه اصلی: فیزیکی، کنترلی، ارتباطی، محاسباتی و مانیتورینگ")
plt.text(8, 10.8, subtitle, ha='center', va='center', fontsize=18, color='darkblue')

# تعریف لایه‌ها
layers = [
    {"name": fa("لایه فیزیکی"), "desc": fa("سنسورها و عملگرها"), "color": "#4e79a7", "height": 1.8},
    {"name": fa("لایه کنترلی"), "desc": fa("PLC و کنترلرهای صنعتی"), "color": "#59a14f", "height": 1.8},
    {"name": fa("لایه ارتباطی"), "desc": fa("دروازه اینترنت اشیا (IoT Gateway)"), "color": "#f28e2c", "height": 1.8},
    {"name": fa("لایه محاسباتی"), "desc": fa("مدل‌های هوش مصنوعی و پردازش داده"), "color": "#e15759", "height": 1.8},
    {"name": fa("لایه مانیتورینگ"), "desc": fa("HMI و سیستم‌های نظارتی"), "color": "#af7aa1", "height": 1.8}
]

# موقعیت‌های لایه‌ها
y_positions = [1.5, 3.5, 5.5, 7.5, 9.5]

# رسم لایه‌ها
for i, layer in enumerate(layers):
    y = y_positions[i]

    # رسم مستطیل لایه
    ax.add_patch(patches.Rectangle((3, y), 10, layer["height"],
                                  facecolor=layer["color"], alpha=0.9, edgecolor='black'))

    # عنوان لایه
    plt.text(8, y + layer["height"]/2 + 0.2, layer["name"],
             ha='center', va='center', fontsize=16, fontweight='bold', color='white')

    # توضیحات لایه
    plt.text(8, y + layer["height"]/2 - 0.2, layer["desc"],
             ha='center', va='center', fontsize=14, color='white')

# افزودن نمادها و آیکون‌ها
symbols = [
    (4, 2.4, '🌡️', fa('سنسور دما'), 1.5),
    (5, 2.4, '📏', fa('سنسور فشار'), 1.5),
    (6, 2.4, '⚗️', fa('سنسور جریان'), 1.5),
    (7, 2.4, '🔄', fa('عملگرها'), 1.5),
    (4, 4.4, '🧠', fa('PLC اصلی'), 2.0),
    (6, 4.4, '💻', fa('کنترلرهای محلی'), 2.0),
    (8, 4.4, '📟', fa('سیستم ایمنی'), 2.0),
    (5, 6.4, '📡', fa('ارتباط بی‌سیم'), 1.5),
    (7, 6.4, '🔌', fa('ارتباط سیمی'), 1.5),
    (9, 6.4, '🔒', fa('امنیت ارتباطات'), 1.5),
    (5, 8.4, '🤖', fa('مدل‌های پیش‌بینی'), 2.0),
    (7, 8.4, '📊', fa('تحلیل داده‌ها'), 2.0),
    (9, 8.4, '🔍', fa('تشخیص ناهنجاری'), 2.0),
    (5, 10.4, '👁️', fa('نمایشگرهای HMI'), 1.5),
    (7, 10.4, '📈', fa('داشبوردهای تحلیلی'), 1.5),
    (9, 10.4, '📲', fa('سیستم‌های موبایل'), 1.5)
]

for x, y, symbol, text, size in symbols:
    plt.text(x, y, symbol, fontsize=size*20, ha='center', va='center')
    plt.text(x, y-0.3, text, ha='center', va='top', fontsize=10)

# رسم فلش‌های ارتباطی بین لایه‌ها
for i in range(len(layers)-1):
    y1 = y_positions[i] + layers[i]["height"]
    y2 = y_positions[i+1]

    # خط اصلی
    ax.plot([8, 8], [y1, y2], 'k-', linewidth=1.5)

    # فلش
    ax.annotate("", xy=(8, y2-0.1), xytext=(8, y1+0.1),
               arrowprops=dict(arrowstyle="->", lw=2, color='black'))

    # توضیح ارتباط
    plt.text(8.5, (y1+y2)/2, fa('جریان داده'), fontsize=10,
             rotation=90, ha='left', va='center',
             bbox=dict(boxstyle='round,pad=0.2', facecolor='white', alpha=0.8))

# افزودن ابرک (Cloud) برای ارتباط خارجی
cloud_x = [12, 12.5, 13, 13.5, 14, 13.5, 13, 12.5]
cloud_y = [8, 7.5, 7.7, 7.5, 8, 8.5, 8.3, 8.5]
ax.plot(cloud_x, cloud_y, 'b-', linewidth=2, alpha=0.7)
plt.text(13, 8, fa('اینترنت صنعتی'), ha='center', va='center', fontsize=12, color='blue')
plt.text(13, 7.5, fa('IIoT'), ha='center', va='center', fontsize=14, fontweight='bold', color='blue')

# اتصال لایه ارتباطی به ابر
ax.plot([11.5, 12], [6.4, 7.8], 'b--', linewidth=1.5)
ax.annotate("", xy=(12, 7.8), xytext=(11.5, 6.4),
           arrowprops=dict(arrowstyle="->", lw=1.5, color='blue', linestyle='--'))

# افزودن توضیحات امنیتی
security = [
    (14, 5, fa('امنیت سایبری'), '#ff6b6b'),
    (14, 4.2, fa('رمزنگاری داده‌ها'), '#ff9e7d'),
    (14, 3.4, fa('احراز هویت'), '#ffd166'),
    (14, 2.6, fa('فایروال صنعتی'), '#06d6a0')
]

for x, y, text, color in security:
    ax.add_patch(patches.Circle((x, y), 0.5, color=color, alpha=0.8))
    plt.text(x, y, text, ha='center', va='center', fontsize=11)

# افزودن جعبه توضیحات پایانی
analysis = fa("""
مزایای کلیدی:
• یکپارچه‌سازی داده‌های فیزیکی و سایبری
• پایش لحظه‌ای تجهیزات و فرآیندها
• پیش‌بینی خرابی با مدل‌های هوش مصنوعی
• کاهش زمان توقف و افزایش ایمنی
• تصمیم‌گیری مبتنی بر داده در زمان واقعی
""")

plt.text(8, 0.8, analysis, ha='center', va='center', fontsize=14,
         bbox=dict(boxstyle='round', facecolor='#e3f2fd', edgecolor='#90caf9', alpha=0.9))

plt.tight_layout()
plt.savefig('cyber_physical_architecture.png', dpi=300, bbox_inches='tight')
plt.show()
```
