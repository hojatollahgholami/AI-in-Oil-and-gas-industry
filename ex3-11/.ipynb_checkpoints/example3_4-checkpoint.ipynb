## فراخوانی کتابخانه های مورد نیاز

```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import pandas as pd
from sklearn.preprocessing import RobustScaler, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
from sklearn.metrics import mean_squared_error, mean_absolute_error
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import seaborn as sns
import statsmodels.api as sm
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.stattools import adfuller
from pmdarima import auto_arima
from xgboost import XGBRegressor
import warnings
warnings.filterwarnings('ignore')
import arabic_reshaper
from bidi.algorithm import get_display

# تابع کمکی برای نمایش صحیح متون فارسی در نمودارها
def fa(text):
    reshaped_text = arabic_reshaper.reshape(text)
    return get_display(reshaped_text)
```

```python
# تنظیمات اولیه
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# تنظیمات نمایش
plt.rcParams['font.family'] = 'Adobe Arabic'  # برای نمایش حروف فارسی
plt.rcParams['font.size'] = 16
```

## بارگزاری داده ها

```python
df = pd.read_excel('feed.xlsx')  
df['timestamp'] = pd.to_datetime(df['timestamp'])
df.set_index('timestamp', inplace=True)

# تعریف ستون های هدف
target_columns = ['Spent%','oBe','H2S', 'CO2', 'RSH']

# تعریف ویژگی های ورودی بر اساس توضیحات شما
feature_columns = [
    'NAR', 'KAN', 'SPG',  # دبی گاز چاه ها
]

# حذف ستون هایی که در داده ها وجود ندارند
feature_columns = [col for col in feature_columns if col in df.columns]
```

## پیش پردازش داده

```python
# بررسی مقادیر و پر کردن آنها
print("مقدار گمشده ها قبل از پر کردن:")
print(df[feature_columns + target_columns].isnull().sum())

# با توجه به نوع مسئله، ستون های ویژگی ها حاوی مقادیر خالی با صفر
df[feature_columns] = df[feature_columns].fillna(0)
# برخی مقادیر ستون های هدف حاوی رشته هستند. لذا تبدیل ستون های غیرعددی به عددی
for col in target_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
# با توجه به نوع مسئله، پر کردن مقادیر ستون های هدف با میانگین متحرک    
df[target_columns] = df[target_columns].interpolate(method='linear') #.bfill()

print(df)
print("مقدار گمشده ها بعد از پر کردن:")
print(df[feature_columns + target_columns].isnull().sum())

# استانداردسازی داده ها
scaler_X = RobustScaler()
scaler_y = RobustScaler()

X_scaled = scaler_X.fit_transform(df[feature_columns])
y_scaled = scaler_y.fit_transform(df[target_columns])
```

## آماده سازی داده برای مدل سازی LSTM

```python
# ایجاد دنباله های زمانی برای LSTM
sequence_length = 72  # استفاده از بخش های 24 ساعته برای پیش بینی
X_seq, y_seq = [], []

for i in range(len(X_scaled) - sequence_length):
    X_seq.append(X_scaled[i:i+sequence_length])
    y_seq.append(y_scaled[i+sequence_length])

X_seq = np.array(X_seq)
y_seq = np.array(y_seq)

# تقسیم داده ها به آموزش و آزمون (80-20)
train_size = int(0.8 * len(X_seq))
X_train, X_test = X_seq[:train_size], X_seq[train_size:]
y_train, y_test = y_seq[:train_size], y_seq[train_size:]

# تبدیل به تنسورهای PyTorch
X_train_tensor = torch.FloatTensor(X_train).to(device)
y_train_tensor = torch.FloatTensor(y_train).to(device)
X_test_tensor = torch.FloatTensor(X_test).to(device)
y_test_tensor = torch.FloatTensor(y_test).to(device)

# تعریف مدل LSTM
class GasLSTMModel(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers, output_size, dropout_rate=0.3):
        super(GasLSTMModel, self).__init__()
        self.hidden_size = hidden_size
        self.num_layers = num_layers
        
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, 
                           batch_first=True, dropout=dropout_rate)
        
        self.dropout = nn.Dropout(dropout_rate)
        self.fc1 = nn.Linear(hidden_size, hidden_size // 2)
        self.fc2 = nn.Linear(hidden_size // 2, output_size)
        self.relu = nn.ReLU()
        
    def forward(self, x):
        h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size).to(device)
        c0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size).to(device)
        
        out, _ = self.lstm(x, (h0, c0))
        out = out[:, -1, :]  # گرفتن آخرین خروجی
        
        out = self.dropout(out)
        out = self.fc1(out)
        out = self.relu(out)
        out = self.fc2(out)
        
        return out

# هایپرپارامترها
input_size = len(feature_columns)
hidden_size = 128
num_layers = 5
output_size = len(target_columns)
num_epochs = 100
learning_rate = 0.001
batch_size = 32

# ایجاد مدل و optimizer
model = GasLSTMModel(input_size, hidden_size, num_layers, output_size).to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate, weight_decay=1e-5)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=10)

# آموزش مدل LSTM
train_losses = []
test_losses = []

# ایجاد DataLoader برای آموزش دسته ای
from torch.utils.data import DataLoader, TensorDataset

train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)

for epoch in range(num_epochs):
    model.train()
    epoch_train_loss = 0
    
    for batch_X, batch_y in train_loader:
        optimizer.zero_grad()
        outputs = model(batch_X)
        loss = criterion(outputs, batch_y)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        epoch_train_loss += loss.item()
    
    # محاسبه loss روی داده آزمون
    model.eval()
    with torch.no_grad():
        test_outputs = model(X_test_tensor)
        test_loss = criterion(test_outputs, y_test_tensor)
    
    train_losses.append(epoch_train_loss / len(train_loader))
    test_losses.append(test_loss.item())
    
    scheduler.step(test_loss)
    
    if (epoch+1) % 10 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {train_losses[-1]:.6f}, Test Loss: {test_losses[-1]:.6f}')

# پیش بینی با مدل LSTM
model.eval()
with torch.no_grad():
    train_predictions = model(X_train_tensor)
    test_predictions = model(X_test_tensor)

# تبدیل پیش بینی ها به مقیاس اصلی
train_predictions = scaler_y.inverse_transform(train_predictions.cpu().numpy())
test_predictions = scaler_y.inverse_transform(test_predictions.cpu().numpy())
y_train_actual = scaler_y.inverse_transform(y_train)
y_test_actual = scaler_y.inverse_transform(y_test)

# 4) منحنی یادگیری (Train/Test Loss در طول Epochها)
plt.figure(figsize=(10, 5))
plt.plot(train_losses, label='Train Loss')
plt.plot(test_losses, label='Test Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Learning Curve')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# ارزیابی مدل LSTM
def evaluate_predictions(actual, predicted, target_names):
    results = {}
    for i, target in enumerate(target_names):
        mse = mean_squared_error(actual[:, i], predicted[:, i])
        mae = mean_absolute_error(actual[:, i], predicted[:, i])
        rmse = np.sqrt(mse)
        
        results[target] = {
            'MSE': mse,
            'MAE': mae,
            'RMSE': rmse,
            'R2': 1 - (mse / np.var(actual[:, i]))
        }
    
    return results

lstm_results = evaluate_predictions(y_test_actual, test_predictions, target_columns)

# ساخت یک دیکشنری برای نمایش بصورت جدول
table_data = []

for target, metrics in lstm_results.items():
    row = {'Target': target}
    row.update(metrics)
    table_data.append(row)

# ایجاد DataFrame
df_results = pd.DataFrame(table_data)

# تنظیم ترتیب ستون‌ها
columns_order = ['Target', 'MSE', 'MAE', 'RMSE', 'R2']
df_results = df_results[columns_order]
print("نتایج ارزیابی مدل LSTM:")
# نمایش جدول
print(df_results.to_string(index=False))
```

## مدل سازی ARIMA

```python
from statsmodels.tsa.stattools import adfuller
from pmdarima import auto_arima
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

# تعریف متغیر برای ذخیره نتایج ARIMA
arima_results = {}
arima_predictions = {}

print("\nتحلیل سری زمانی با ARIMA:")

# تابع بررسی ایستایی
def check_stationarity(series, name):
    result = adfuller(series.dropna())
    print(f'آزمون دیکی-فولر برای {name}:')
    print(f'آماره ADF: {result[0]:.4f}')
    print(f'p-value: {result[1]:.4f}')
    if result[1] <= 0.05:
        print("داده ایستا است (H0 رد شد)")
    else:
        print("داده غیرایستا است (H0 رد نشد)")
    print()

# تحلیل برای هر متغیر هدف
for i, target in enumerate(target_columns):
    print(f"تحلیل برای {target}:")
    
    # داده ها
    train_size_arima = int(0.8 * len(df))
    train_data = df[target].iloc[:train_size_arima]
    test_data = df[target].iloc[train_size_arima:]
    
    # بررسی ایستایی
    check_stationarity(train_data, target)
    
    # انتخاب بهترین مدل
    model_arima = auto_arima(train_data,
                             seasonal=False,   # ← اگر داده ها روزانه/ساعتی هستند و فصل پذیری داری بذار True
                             stepwise=True,
                             suppress_warnings=True,
                             error_action='ignore',
                             trace=False)
    
    model_fit = model_arima.fit(train_data)
    
    # پیش بینی روی داده تست
    forecast = model_fit.predict(n_periods=len(test_data))
    arima_predictions[target] = forecast
    
    # ذخیره نتایج عددی
    mse = mean_squared_error(test_data, forecast)
    mae = mean_absolute_error(test_data, forecast)
    rmse = np.sqrt(mse)
    r2 = r2_score(test_data, forecast)
    
    arima_results[target] = {
        'MSE': mse,
        'MAE': mae,
        'RMSE': rmse,
        'R2': r2
    }
    
    print(f"نتایج ARIMA برای {target}:")
    for metric, value in arima_results[target].items():
        print(f"  {metric}: {value:.4f}")
    print("-"*40)
```

## مدل سازی با XGBoost

```python
# دیکشنری برای ذخیره نتایج
xgb_results = {}
xgb_predictions = {}

# تعداد lag ها
n_lags = 5

for target in target_columns:
    print(f"\n مدل سازی XGBoost برای {target}")
    
    df_tmp = df.copy()
    
    # ایجاد ویژگی های تاخیری
    for lag in range(1, n_lags+1):
        df_tmp[f"lag_{lag}"] = df_tmp[target].shift(lag)
    
    # ویژگی های آماری
    df_tmp["rolling_mean_3"] = df_tmp[target].rolling(window=3).mean()
    df_tmp["rolling_std_3"]  = df_tmp[target].rolling(window=3).std()
    
    # حذف NaN
    df_tmp = df_tmp.dropna()
    
    # X و y
    X = df_tmp.drop(columns=[target])
    y = df_tmp[target]
    
    # تقسیم train/test
    split = int(len(df_tmp) * 0.8)
    X_train, X_test = X.iloc[:split], X.iloc[split:]
    y_train, y_test = y.iloc[:split], y.iloc[split:]
    
    # مدل
    model = XGBRegressor(
        n_estimators=500,
        learning_rate=0.05,
        max_depth=5,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42
    )
    
    # آموزش
    model.fit(X_train, y_train)
    
    # پیش‌بینی
    y_pred = model.predict(X_test)
    
    # ذخیره پیش‌بینی‌ها
    xgb_predictions[target] = y_pred
    
    # ارزیابی
    mse = mean_squared_error(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    
    xgb_results[target] = {"MSE": mse, "MAE": mae, "RMSE": rmse, "R2": r2}
    
    # نمایش نتایج
    print(f"RMSE: {rmse:.4f}, R²: {r2:.4f}")
    
    # رسم نمودار
    plt.figure(figsize=(12,5))
    plt.plot(y_test.values, label=fa("واقعی"), linewidth=2)
    plt.plot(y_pred, label=fa("پیش بینی XGBoost"), linewidth=2)
    plt.legend()
    plt.title(fa(f"مقایسه واقعی و پیش بینی برای {target} با XGBoost"))
    plt.show()

# جدول نهایی
xgb_df = pd.DataFrame(xgb_results).T
print("\n نتایج نهایی XGBoost برای همه اهداف:")
display(xgb_df)
```

XGBoost بصورت ذاتی یک مدل پویا مثل ARIMA یا State-Space نیست، چون دینامیک زمان را مستقیماً مدل نمی کند.
اما وقتی ویژگی های زمانی (lag, rolling, trend, seasonal) به مدل بدهیم، XGBoost .می تواند رفتار پویا را یاد بگیرد و در عمل مثل یک مدل سری زمانی عمل کند.
بنابراین:

XGBoost → ذاتاً پویا نیست، ولی با مهندسی ویژگی ها می تواند بصورت پویا رفتار کند.

 چون الگوریتم یادگیری غیرخطی XGBoost، وابستگی های زمانی را از لگ ها یاد گرفته.

## مقایسه خطاها

```python
# --- ساخت time_index متناظر با y_seq/y_test ---
# طول کل دنباله ها = len(X_seq)  (یعنی len(X_scaled) - sequence_length)
# y_seq[i] معادل timestamp در df.index[i + sequence_length]
seq_start_index = sequence_length
# اندیس شروع برای y_test در y_seq برابر train_size است
y_test_start_idx_in_df = seq_start_index + train_size
# زمان های مربوط به y_test_actual:
time_index = df.index[y_test_start_idx_in_df : y_test_start_idx_in_df + len(y_test_actual)]
# اگر خواستید همه y_test_actual را بگیرید:
time_index = pd.to_datetime(time_index)

# دیکشنری برای ذخیره خطاها
comparison_errors = {target: {} for target in target_columns}

for i, target in enumerate(target_columns):
    print(f"\n مقایسه مدل ها برای {target}")
    
    # --- داده ها ---
    x = np.array(time_index)
    y_actual = np.array(y_test_actual)[:, i]
    y_pred_lstm = np.array(test_predictions)[:, i]
    y_pred_arima = np.array(arima_predictions[target])
    y_pred_xgb = np.array(xgb_predictions[target])   
    
    # هماهنگ کردن طول پیش بینی ها با داده واقعی
    min_len = min(len(x), len(y_pred_arima), len(y_actual), len(y_pred_lstm))
    x = x[-min_len:]
    y_actual = y_actual[:min_len]
    y_pred_lstm = y_pred_lstm[:min_len]
    y_pred_arima = y_pred_arima[:min_len]
    y_pred_xgb = y_pred_xgb[:min_len]

    # --- محاسبه خطاها ---
    mse_lstm = mean_squared_error(y_actual, y_pred_lstm)
    mse_arima = mean_squared_error(y_actual, y_pred_arima[:len(y_actual)])
    mse_xgb = mean_squared_error(y_actual, y_pred_xgb)
    
    rmse_lstm = np.sqrt(mse_lstm)
    rmse_arima = np.sqrt(mse_arima)
    rmse_xgb = np.sqrt(mse_xgb)
    
    comparison_errors[target] = {
        "LSTM": rmse_lstm,
        "ARIMA": rmse_arima,
        "XGBoost": rmse_xgb
    }
    
    # --- نمودار سری زمانی ---
    plt.figure(figsize=(14,6))
    plt.plot(x, y_actual, label=fa("واقعی"), linewidth=2)
    plt.plot(x, y_pred_lstm, label="LSTM", linestyle="--")
    plt.plot(x, y_pred_arima[:len(y_actual)], label="ARIMA", linestyle="--")
    plt.plot(x, y_pred_xgb, label="XGBoost", linestyle="--")
    plt.title(fa(f"مقایسه پیش بینی مدل ها برای {target}"))
    plt.legend()
    plt.show()
    
    # --- نمودار میله ای خطاها ---
    plt.figure(figsize=(6,4))
    plt.bar(comparison_errors[target].keys(), comparison_errors[target].values(),
            color=["orange", "blue", "green"])
    plt.ylabel("RMSE")
    plt.title(fa(f"مقایسه خطاهای مدل ها برای {target}"))
    plt.show()

# در پایان می‌توان یک جدول از خطاها هم ساخت
import pandas as pd
error_df = pd.DataFrame(comparison_errors).T
print("\n📊 جدول خطاها (RMSE):")
display(error_df)
```
